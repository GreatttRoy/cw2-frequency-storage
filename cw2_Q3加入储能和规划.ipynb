### 1. EFR 模型与单母线系统

按照论文的设定，事故发生后，EFR 功率线性上升，在交付时间内达到安排容量，随后保持不变：

$$
p_B(t)=
\begin{cases}
\dfrac{R}{T_s}t, & 0\leq t<T_s,\\[6pt]
R, & t\geq T_s.
\end{cases}
$$

将其接入第（1）问的单母线系统，并保留原调速器模型：

$$
m\dot w=p_m-p_{\ell,0}-dw+p_B.\tag{1}
$$

$$
\tau_g\dot p_m+p_m=-k_gw,
\qquad \Delta f=f_0w.
$$

所有功率均采用标幺值。初始状态取

$$
w(0)=0,\qquad p_m(0)=0.
$$



净输入：储能出力（+）、负荷阶跃（−）、调速器反馈（+）→ 摆动环节 → 标幺频差 → 乘以额定频率得到 Hz 偏差。

### 2. 分段功率的拉普拉斯变换

对 EFR 功率分段积分：

$$
P_B(s)=
\int_0^{T_s}\frac{Rt}{T_s}e^{-st}\,dt
+\int_{T_s}^{\infty}Re^{-st}\,dt.
$$

爬坡阶段的积分为

$$
\int_0^{T_s}\frac{Rt}{T_s}e^{-st}\,dt
=\frac{R}{T_ss^2}
\left[1-(1+T_ss)e^{-T_ss}\right].
$$

恒功率阶段的积分为

$$
\int_{T_s}^{\infty}Re^{-st}\,dt
=\frac{R}{s}e^{-T_ss}.
$$

合并后得到

$$
P_B(s)=\frac{R(1-e^{-T_ss})}{T_ss^2}.
$$

定义系统特征多项式：

$$
q(s)=m\tau_gs^2+(m+d\tau_g)s+d+k_g.
$$

频率偏差的拉普拉斯表达式为

$$
\Delta F(s)=
\frac{f_0(1+\tau_gs)}{q(s)}
\left[
-\frac{p_{\ell,0}}{s}
+\frac{R(1-e^{-T_ss})}{T_ss^2}
\right].
$$

### 3. 分段频率响应解析式

对于欠阻尼系统，定义

$$
\alpha=\frac{m+d\tau_g}{2m\tau_g},
\qquad
\beta=
\frac{\sqrt{4m\tau_g(d+k_g)-(m+d\tau_g)^2}}
{2m\tau_g},
$$

$$
\lambda=-\alpha+i\beta,
\qquad
A=\frac{1+\tau_g\lambda}{2im\tau_g\beta}.
$$

再定义

$$
H(t)=2\operatorname{Re}
\left[A\frac{e^{\lambda t}-1}{\lambda}\right],
$$

$$
J(t)=2\operatorname{Re}\left[
A\left(\frac{e^{\lambda t}-1}{\lambda^2}
-\frac{t}{\lambda}\right)\right].
$$

反变换后，频率偏差为

$$
\Delta f(t)=
\begin{cases}
f_0\left[-p_{\ell,0}H(t)+\dfrac{R}{T_s}J(t)\right],
& 0\leq t<T_s,\\[8pt]
f_0\left[-p_{\ell,0}H(t)
+\dfrac{R}{T_s}\bigl(J(t)-J(t-T_s)\bigr)\right],
& t\geq T_s.
\end{cases}
$$

其中，第一段对应 EFR 爬坡，第二段对应 EFR 恒功率输出。



### 4. 初始 RoCoF

由于 EFR 在事故瞬间的输出为零：

$$
p_B(0^+)=0,
$$

初始 RoCoF 为

$$
\dot{\Delta f}(0^+)=-\frac{f_0p_{\ell,0}}{m}.
$$

因此，在该爬坡模型下，增加储能单元数量不改变初始 RoCoF。



### 5. 准稳态频率偏差

若 EFR 持续保持输出，系统达到平衡时：

$$
\Delta f_{\mathrm{ss}}
=-\frac{f_0(p_{\ell,0}-R)}{d+k_g}.
$$

当总 EFR 容量不超过负荷扰动时，增加单元数量会减小准稳态偏差的绝对值。该式以 EFR 持续输出为前提；若储能退出或恢复充电，需要重新计算退出后的频率响应。

实际频率为 $f_{\rm sys}=f_0+\Delta f$，本例稳态为 49.785714 Hz，不含二次调频。

### 6. 频率最低点及其发生时刻

#### （1）最低点发生在 EFR 爬坡阶段

若最低点发生在爬坡阶段，其时刻由以下方程确定：

$$
2\operatorname{Re}\left\{
A\left[
\left(\frac{R}{T_s\lambda}-p_{\ell,0}\right)e^{\lambda t_1}
-\frac{R}{T_s\lambda}
\right]\right\}=0,
\qquad 0<t_1<T_s.
$$

对应的频率偏差为

$$
\Delta f_1=f_0\left[
-p_{\ell,0}H(t_1)+\frac{R}{T_s}J(t_1)
\right].
$$

此时刻方程一般不能用初等函数显式表示，需数值求根。

#### （2）最低点发生在 EFR 恒功率阶段

定义

$$
z=A\left[
-p_{\ell,0}
+\frac{R}{T_s}\frac{1-e^{-\lambda T_s}}{\lambda}
\right],
\qquad
\phi=\arg z.
$$

当上式中的复数不为零时，恒功率阶段第一个局部最低点为

$$
t_2=\frac{3\pi/2-\phi+2\pi n}{\beta},
$$

$$
n=\left\lceil
\frac{\beta T_s+\phi-3\pi/2}{2\pi}
\right\rceil.
$$

对应的频率偏差为

$$
\Delta f_2=
-\frac{f_0(p_{\ell,0}-R)}{d+k_g}
-\frac{2f_0\beta|z|}{\alpha^2+\beta^2}
e^{-\alpha t_2}.
$$

最终比较爬坡阶段候选最低点、分段边界和恒功率阶段的最低点，即可确定整个响应的频率最低点及其发生时刻。增加储能单元数量会改变上述最低点公式中的总 EFR 容量；最低点时刻的变化方向需结合系统参数计算。

### 解析响应与独立时域校验

```python
# 静默当前内核反复报告的矩阵乘法和内嵌图表提示。
import warnings
warnings.filterwarnings("ignore", message=r"(divide by zero|overflow|invalid value) encountered in matmul$", category=RuntimeWarning)
warnings.filterwarnings("ignore", message=r"Consider using IPython.display.IFrame instead", category=UserWarning)

import control as ct  # python-control：传递函数、反馈连接和阶跃响应

import sys
print(f"Python: {sys.executable} | control: {ct.__version__}")
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display

base = dict(m=10., d=1., k=20., tau=5.)
f0, P = 50., 0.10

R, Ts = .01, 1.

def simulate(m, d, k, tau, dt=0.001):
    swing = ct.tf([1], [m, d])
    governor = ct.tf([k], [tau, 1])
    system = f0 * ct.feedback(swing, governor)
    end = 12 / min(-ct.poles(system).real)
    t = np.arange(int(np.ceil(end / dt)) + 1) * dt
    t, y = ct.forced_response(system, T=t, U=-P+R*np.minimum(t/Ts,1))
    y = np.asarray(y)
    i = y.argmin()
    metrics = [(-3*y[0]+4*y[1]-y[2])/(2*dt), y[-round(1/dt):].mean(), y[i], t[i]]
    return t, y, metrics

from scipy.optimize import brentq
lam = -.15 + 1j*np.sqrt(.42-.15**2)
coef = (1+5*lam)/(100j*lam.imag)
def H(t):
    return 2*(coef*np.expm1(lam*t)/lam).real
def J(t):
    return 2*(coef*(np.expm1(lam*t)/lam**2-t/lam)).real
def analytical(t, load=P, reserve=R):
    return f0*(-load*H(t)+reserve/Ts*(J(t)-J(np.maximum(t-Ts,0))))
def slope(t):
    return f0*(-2*P*(coef*np.exp(lam*t)).real+R/Ts*(H(t)-H(max(t-Ts,0))))
z = coef*(-P+R/Ts*(-np.expm1(-lam*Ts))/lam)
phase = np.angle(z)
tn = (1.5*np.pi-phase+2*np.pi*np.ceil((lam.imag*Ts+phase-1.5*np.pi)/(2*np.pi)))/lam.imag
knots=np.linspace(0,Ts,101)
candidates=[0.,Ts,tn]+[brentq(slope,left,right) for left,right in zip(knots,knots[1:]) if slope(left)*slope(right)<0]
tn=min(candidates,key=analytical)
exact=np.array([-f0*P/10,-f0*(P-R)/21,analytical(tn),tn])
assert exact[2]<exact[1]
t, y, measured = simulate(**base)
labels = ['RoCoF (Hz/s)', '准稳态偏差 (Hz)', '最低点偏差 (Hz)', '最低点时刻 (s)']
print("表 3")
display(pd.DataFrame({'仿真值': measured, '解析值': exact, '绝对误差': abs(measured-exact)}, index=labels).style.format('{:.9f}'))
_, _, finer = simulate(**base, dt=0.0005)
assert abs(finer[3]-tn) <= 0.0005 and abs(measured[3]-tn) <= 0.001
fig, ax = plt.subplots()
ax.plot(t, y, label='python-control')
ax.plot(t, analytical(t), '--', label='Analytical')
ax.plot(tn, analytical(tn), 'ro', label='Nadir')
ax.axhline(exact[1], color='gray', ls=':', label='Quasi-steady state')
ax.set(xlabel='Time (s)', ylabel='Frequency deviation (Hz)')
ax.legend()
plt.show()
print("图 1")
assert np.max(abs(y-analytical(t)))<1e-7
assert abs(slope(Ts-1e-8)-slope(Ts+1e-8))<1e-7
for load,reserve in [(P,0.),(R,R)]:
    sys=ct.tf([50*5,50],[50,15,21])
    _, check=ct.forced_response(sys,T=t,U=-load+reserve*np.minimum(t/Ts,1))
    assert np.max(abs(check-analytical(t,load,reserve)))<1e-7 and check.min()<0
```

## 容量与谷值边界分层采样
每档扰动抽取内部、面及轴上的容量配比。固定配比时总容量加倍，频差减半，因此可从一次动态计算直接定位 49.50 Hz 两侧的容量。

```python
# 静默当前内核反复报告的矩阵乘法和内嵌图表提示。
import warnings
warnings.filterwarnings("ignore", message=r"(divide by zero|overflow|invalid value) encountered in matmul$", category=RuntimeWarning)
warnings.filterwarnings("ignore", message=r"Consider using IPython.display.IFrame instead", category=UserWarning)

import numpy as np
import pandas as pd
import pyarrow as pa
import pyarrow.parquet as pq
from pathlib import Path
from tqdm import tqdm as Tqdm
from IPython.display import display, HTML
from html import escape
from io import StringIO
import control as ct

m_i = np.array([14., 8., 4.])
d_i = np.array([0.8, 1.1, 1.5])
k_i = np.array([15., 23., 30.])
tau_i = np.array([7., 3., 0.8])
threshold = np.array([0.50, 0.30, 0.50])

delivery_time = 0.5
unit_cost = 0.5516
def storage_bound(load):
    return int(np.floor(100*(load+.4536)+1e-9))
```

对每个配比展开单位负荷阶跃的模态；积分得到斜坡响应，按实际扰动与固定储能容量叠加后再求驻点。同比扩大装机容量只缩小频差，不改变驻点时间。

```python
def response_batch(S, load, count):
    """1 p.u. 阶跃，输出正的频率下降量。状态为频率和三个调速器功率增量。"""
    Ts = delivery_time
    R = np.broadcast_to(.01*np.asarray(count),len(S))
    M = (S * m_i).sum(axis=1) / 100
    D = (S * d_i).sum(axis=1) / 100
    K = S * k_i / 100
    A = np.zeros((len(S), 4, 4))
    A[:, 0, 0] = -D / M
    A[:, 0, 1:] = 1 / M[:, None]
    A[:, 1:, 0] = -K / tau_i
    A[:, [1, 2, 3], [1, 2, 3]] = -1 / tau_i
    poles, vectors = np.linalg.eig(A)
    steady = 50 / (D + K.sum(axis=1))
    equilibrium = np.column_stack((-steady / 50, K * steady[:, None] / 50))
    weights = np.linalg.solve(vectors, equilibrium[..., None])[..., 0]
    modes = 50 * vectors[:, 0, :] * weights
    assert np.all(poles.real < 0)
    best=np.maximum(0,(load-R)*steady)
    times=np.where(best>0,np.inf,0.)
    for ramp in [True,False]:
        if ramp:
            coeff=modes*(load-R[:,None]/(Ts*poles))
            constant=load*steady+(R[:,None]/Ts*modes/poles).sum(axis=1).real
            linear=-R/Ts*steady
        else:
            coeff=modes*(load*np.exp(poles*Ts)-R[:,None]/Ts*np.expm1(poles*Ts)/poles)
            constant=(load-R)*steady
            linear=np.zeros_like(steady)
        step=min(.1,.18/max(1,np.max(abs(poles.imag))))
        step=Ts/np.ceil(Ts/step)
        elapsed=0.
        term=coeff.copy()
        while True:
            value=constant+linear*elapsed+term.sum(axis=1).real
            better=value>best
            best=np.maximum(value,best)
            times=np.where(better,elapsed+(0 if ramp else Ts),times)
            if ramp and elapsed>=Ts-1e-12: break
            if not ramp and np.all(constant+abs(term).sum(axis=1)<=best+1e-12): break
            nxt=term*np.exp(poles*step)
            ix=np.flatnonzero(((term*poles).sum(axis=1).real+linear>0)&((nxt*poles).sum(axis=1).real+linear<=0))
            if ix.size:
                lo=np.zeros(ix.size); hi=lo+step
                for _ in range(22):
                    mid=(hi+lo)/2
                    rising=(term[ix]*poles[ix]*np.exp(poles[ix]*mid[:,None])).sum(axis=1).real+linear[ix]>0
                    lo=np.where(rising,mid,lo); hi=np.where(rising,hi,mid)
                val=constant[ix]+linear[ix]*(elapsed+mid)+(term[ix]*np.exp(poles[ix]*mid[:,None])).sum(axis=1).real
                better=val>best[ix]
                times[ix[better]]=elapsed+mid[better]+(0 if ramp else Ts)
                best[ix]=np.maximum(best[ix],val)
            elapsed+=step
            term=nxt
    return best,times
```

### 随机总容量与边界定向采样
总量上限 240 MW。全局样本覆盖宽扰动范围，定向样本在边界与容量上限相交的可行区域生成。

```python
rng=np.random.default_rng(42)
parts=[]
for level in range(1,101):
    load=level/100
    mix=rng.dirichlet(np.ones(3),320)
    for j in range(3):
        mix[40*j:40*(j+1),j]=0
    mix/=mix.sum(axis=1,keepdims=True)
    mix=np.vstack([mix,np.eye(3)])
    counts=rng.integers(0,storage_bound(load)+1,len(mix))
    counts[::5]=0
    counts[1::5]=level
    counts[2::5]=rng.integers(0,level+1,len(counts[2::5]))
    drop=response_batch(100*mix,load,counts)[0]
    def add(total,mask):
        parts.append(np.column_stack([mix[mask]*total[mask,None],np.full(mask.sum(),load),counts[mask],50-100*drop[mask]/total[mask]]))
    total=rng.uniform(5,240,len(mix))
    add(total,np.ones(len(mix),bool))
    for side in [-1,1]:
        for _ in range(4):
            total=100*drop/(.5+side*rng.uniform(.001,.05,len(mix)))
            add(total,(total>0)&(total<=240))
features=["S1_MW","S2_MW","S3_MW","P_pu","N"]
target="f_nadir_hz"
sample=pd.DataFrame(np.vstack(parts),columns=features+[target]).drop_duplicates()
sample["N"]=sample["N"].astype(int)
sample.to_parquet("cw2_joint_samples.parquet",index=False)
print(len(sample),((sample[target]>=49.45)&(sample[target]<49.5)).sum(),sample[target].between(49.5,49.55).sum())
```

### 独立积分复核

```python
from scipy.integrate import solve_ivp
inertias,damping,gains,lags=m_i,d_i,k_i,tau_i
def frequency_metrics(S,P,count):
    R,Ts=.01*count,delivery_time
    M = np.sum(inertias*S)/100
    D = np.sum(damping*S)/100
    K = gains*S/100
    def dynamics(t,x):
        return np.r_[(x[1:].sum()+R*min(t/Ts,1)-P-D*x[0])/M,(-K*x[0]-x[1:])/lags]
    def turning_point(t,x):
        return (x[1:].sum()+R*min(t/Ts,1)-P-D*x[0])/M
    turning_point.direction = 1
    result = solve_ivp(dynamics,(0,80),np.zeros(4),dense_output=True,
                       rtol=1e-12,atol=1e-14,max_step=.25)
    if not result.success:
        raise RuntimeError(result.message)
    steady = -50*(P-R)/(D+K.sum())
    grid=np.arange(0,80.001,.05)
    samples=50*result.sol(grid)[0]
    from scipy.optimize import minimize_scalar
    troughs=np.flatnonzero((samples[1:-1]<samples[:-2])&(samples[1:-1]<=samples[2:]))+1
    candidates=[steady,samples.min(),*(minimize_scalar(lambda t:50*result.sol(t)[0],bounds=(grid[j-1],grid[j+1]),method='bounded').fun for j in troughs)]
    return -50*P/M, steady, min(candidates)

for row in sample.sample(10,random_state=1).to_numpy():
    assert abs(50+frequency_metrics(row[:3],row[3],row[4])[2]-row[5])<2e-6

trial=np.array([50.,30.,20.])
without=frequency_metrics(trial,.03,0)
five=frequency_metrics(trial,.03,5)
assert without[0]==five[0] and without[1]<0<five[1]
assert np.isclose(5*unit_cost,2.758) and np.isclose(.01*5*100,5)
assert np.isclose(50-response_batch(trial[None,:],.03,0)[0][0],50+without[2],atol=1e-6)
```

## 八种 ReLU 谷值拟合器
输入 $h^{(0)}=(S_1/100,S_2/100,S_3/100,P,N/100)$。逐层计算 $q^{(l)}=W_l^\top h^{(l-1)}+a_l,\ h^{(l)}=\max(0,q^{(l)})$。输出 $\widehat f=50+b+v^\top h^{(L)}$，以实际谷值的均方误差训练。

```python
from sklearn.model_selection import GroupShuffleSplit
from sklearn.neural_network import MLPRegressor
from threadpoolctl import threadpool_limits
from copy import deepcopy
X=sample[features].to_numpy()/[100,100,100,1,100]
y=sample[target].to_numpy()-50
groups=pd.util.hash_pandas_object(sample[features[:3]+["N"]],index=False)
splitter=GroupShuffleSplit(n_splits=1,test_size=.3,random_state=42)
train,hold=next(splitter.split(X,groups=groups))
left,right=next(GroupShuffleSplit(n_splits=1,test_size=.5,random_state=43).split(X[hold],groups=groups.iloc[hold]))
validation,test=hold[left],hold[right]
```

### 相同容量划分与训练预算
八种结构均按边界验证集均方误差保留最佳轮次。新增三隐藏层结构用于表 9 的拟合精度比较。

```python
models={}
accuracy_models = {}
history_64 = []
with threadpool_limits(limits=1):
    for name,shape in [("16",(16,)),("32",(32,)),("16+16",(16,16)),("32+32",(32,32)),("64+64",(64,64)),("32+32+32",(32,)*3),("64+64+64",(64,)*3),("100+100+100",(100,)*3)]:
        model=MLPRegressor(hidden_layer_sizes=shape,batch_size=256,random_state=42,learning_rate_init=.002)
        best_loss=np.inf
        near=validation[np.abs(y[validation]+.5)<=.05]
        for epoch in range(100):
            model.partial_fit(X[train],y[train])
            loss=np.mean((model.predict(X[near])-y[near])**2)
            if name == "64+64":
                train_mse = np.mean((model.predict(X[train])-y[train])**2)
                history_64.append((epoch+1, train_mse, loss))
            if loss<best_loss:
                best_loss=loss
                accuracy_models[name]=deepcopy(model)
        if len(shape) <= 2:
            models[name] = accuracy_models[name]
```

### 谷值频率回归约束
ReLU 网络嵌入优化的完整线性化见式（21″a）至（21″h）。


### 含 EFR 的容量优化模型
$$\min_{S,N}\ \sum_i c_iS_i+c_BN \tag{18″}$$
$$|\dot{\Delta f}(0^+)|=\frac{f_0P}{M}\le R_{\max}\tag{19″}$$
$$|\Delta f_{\rm ss}|=\frac{f_0|P-r_BN|}{D+\sum_iK_i}\le F_{\rm ss,max}\tag{20″}$$
$$\widehat f_{\rm nadir}(S_1,S_2,S_3,P,N)=f_0+b+\sum_jv_jh_j\ge f_{\min}\tag{21″}$$
$$S_i\ge0,\qquad\sum_i S_i\le S_{\max},\quad N\in\mathbb Z_{\ge0}.\tag{22″}$$
仅式（21″）使用回归近似；由准稳态约束及 $D+\sum_iK_i\le G_{\max}$ 可得到符号表中的 $N_{\max}$。

### 谷值频率回归约束的大 M 线性化

式（21″）的 ReLU 网络通过以下约束精确嵌入。令网络有 $L$ 个隐藏层，第 $l$ 层神经元集合为 $\mathcal J_l$，输入为
$$h^{(0)}=(S_1/S_{\rm B},S_2/S_{\rm B},S_3/S_{\rm B},P,N/N_{\rm sc})^\top,\quad N_{\max}=\left\lfloor(P+F_{\rm ss,max}G_{\max}/f_0)/r_B\right\rfloor. \tag{21″a}$$
这里 $N_{\max}$ 是对固定扰动 $P$ 推导出的储能台数有效上界，用于限定优化变量和计算大 M 界，并非固定安装数量或独立的场站容量上限。由准稳态约束 $|P-r_BN|\le F_{\rm ss,max}(D+\sum_iK_i)/f_0$ 及 $D+\sum_iK_i\le G_{\max}$ 得到该上界；实际容量组合还必须满足完整准稳态约束。

采样时每档 $P$ 在 $0$ 至 $N_{\max}$ 的整数范围随机抽取台数，并额外安排 $N=0$、$N=P/r_B$ 及不大于 $P/r_B$ 的台数，以覆盖无储能、恰好补偿及不同响应水平。当前 $P$ 为 0.01 至 1.00 p.u.，对应上界分别为 46 至 145 台；随机抽取不保证每档都覆盖所有整数。

每层仿射输入及输出为
$$q_j^{(l)}=a_j^{(l)}+\sum_{i\in\mathcal J_{l-1}}w_{ij}^{(l)}h_i^{(l-1)},\qquad h_j^{(l)}=\max(0,q_j^{(l)}). \tag{21″b}$$
固定 $P$ 并记 $c_j=a_j^{(1)}+w_{4j}^{(1)}P$，第一层的有效界为
$$\begin{aligned}
L_j^{(1)}&=c_j+\frac{S_{\max}}{S_{\rm B}}\min(0,w_{1j}^{(1)},w_{2j}^{(1)},w_{3j}^{(1)})+\frac{N_{\max}}{N_{\rm sc}}\min(0,w_{5j}^{(1)}),\\
U_j^{(1)}&=c_j+\frac{S_{\max}}{S_{\rm B}}\max(0,w_{1j}^{(1)},w_{2j}^{(1)},w_{3j}^{(1)})+\frac{N_{\max}}{N_{\rm sc}}\max(0,w_{5j}^{(1)}).
\end{aligned}\tag{21″c}$$
对后续层，令 $\underline h_i^{(l-1)}=\max(0,L_i^{(l-1)})$、$\overline h_i^{(l-1)}=\max(0,U_i^{(l-1)})$，以及 $w^+=\max(w,0)$、$w^-=\min(w,0)$，有
$$\begin{aligned}
L_j^{(l)}&=a_j^{(l)}+\sum_i\left(w_{ij}^{(l)+}\underline h_i^{(l-1)}+w_{ij}^{(l)-}\overline h_i^{(l-1)}\right),\\
U_j^{(l)}&=a_j^{(l)}+\sum_i\left(w_{ij}^{(l)+}\overline h_i^{(l-1)}+w_{ij}^{(l)-}\underline h_i^{(l-1)}\right),\quad l\ge2.
\end{aligned}\tag{21″d}$$
对于 $L_j^{(l)}<0<U_j^{(l)}$ 的神经元，引入 $z_j^{(l)}\in\{0,1\}$，实施
$$\begin{aligned}
h_j^{(l)}&\ge0,\quad h_j^{(l)}\ge q_j^{(l)},\\
h_j^{(l)}&\le U_j^{(l)}z_j^{(l)},\quad
h_j^{(l)}\le q_j^{(l)}-L_j^{(l)}(1-z_j^{(l)}).
\end{aligned}\tag{21″e}$$
对于符号固定的神经元，无需二进制变量：
$$h_j^{(l)}=0\quad(U_j^{(l)}\le0),\qquad h_j^{(l)}=q_j^{(l)}\quad(L_j^{(l)}\ge0). \tag{21″f}$$
输出层与频率约束分别为
$$\widehat f_{\rm nadir}=f_0+b+\sum_{j\in\mathcal J_L}v_jh_j^{(L)}, \tag{21″g}$$
$$\widehat f_{\rm nadir}\ge f_{\min}. \tag{21″h}$$
上述约束与线性成本、容量约束、两条准稳态线性不等式及整数台数 $N$ 构成混合整数线性规划。大 M 形式精确表示已训练的 ReLU 网络，动态谷值仍需仿真复核。

### Gurobi 容量优化与扰动扫描

每次求解固定一个 $P$，重算神经元界，按式（22″）允许容量连续取值。扫描 0.01 至 1.00 p.u.，步长 0.01。表格展示最小可行扰动、题设工况及最大可行扰动。曲线仅连接有最优解的采样点。频率最低点由原模型的时域驻点与稳态值复核。

```python
import numpy as np
import pandas as pd
import gurobipy as gp
from gurobipy import GRB
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp
from IPython.display import display

costs = np.array([0.75,1.05,1.65])
inertias = np.array([14.,8.,4.])
damping = np.array([.8,1.1,1.5])
gains = np.array([15.,23.,30.])
lags = np.array([7.,3.,.8])

def solve_capacity(P,net, report=False):
    with gp.Model() as model:
        model.Params.OutputFlag=0
        model.Params.MIPGap=1e-7
        S=model.addMVar(3,lb=0)
        limit=storage_bound(P)
        count=model.addMVar((),lb=0,ub=limit,vtype=GRB.INTEGER,name="N")
        model.setObjective(costs@S+unit_cost*count)
        model.addConstr(S.sum()<=240)
        model.addConstr(inertias@S>=10000*P)
        for sign in [-1,1]:
            model.addConstr(.3*((damping+gains)@S)>=sign*5000*(P-.01*count))
        for layer,(W,bias) in enumerate(zip(net.coefs_[:-1],net.intercepts_[:-1])):
            if layer==0:
                offset=bias+P*W[3]
                expr=W[:3].T@S/100+offset+W[4]*count/100
                lower=offset+2.4*np.minimum(0,W[:3].min(axis=0))+limit*np.minimum(W[4],0)/100
                upper=offset+2.4*np.maximum(0,W[:3].max(axis=0))+limit*np.maximum(W[4],0)/100
            else:
                expr=W.T@h+bias
                pos,neg=np.maximum(W,0),np.minimum(W,0)
                lower,upper=bias+lo@pos+hi@neg,bias+hi@pos+lo@neg
            lo,hi=np.maximum(lower,0),np.maximum(upper,0)
            h=model.addMVar(len(bias),lb=lo,ub=hi)
            for j in range(len(bias)):
                if upper[j]<=0: continue
                if lower[j]>=0: model.addConstr(h[j]==expr[j])
                else:
                    z=model.addVar(vtype=GRB.BINARY)
                    model.addConstr(h[j]>=expr[j])
                    model.addConstr(h[j]<=upper[j]*z)
                    model.addConstr(h[j]<=expr[j]-lower[j]*(1-z))
        model.addConstr(50+net.intercepts_[-1].item()+net.coefs_[-1].ravel()@h>=49.5)
        model.update()
        model.optimize()
        if report:
            stats = {
                "问题类型": "MILP" if model.IsMIP and not model.IsQP and not model.IsQCP else "其他",
                "线性约束行数": model.NumConstrs,
                "变量列数": model.NumVars,
                "连续变量": model.NumVars-model.NumIntVars,
                "整数变量（含二进制）": model.NumIntVars,
                "二进制变量": model.NumBinVars,
                "非零系数": model.NumNZs,
                "求解方法": "Gurobi MIP，预处理及分支定界/割平面；根节点方法自动选择",
                "Gurobi 版本": ".".join(map(str, gp.gurobi.version())),
                "求解状态": model.Status,
                "求解时间（秒）": model.Runtime,
                "搜索节点数": model.NodeCount,
                "MIPGap": model.MIPGap if model.SolCount else np.nan,
            }
            display(pd.DataFrame([stats]))

        if model.Status==GRB.INFEASIBLE: return None
        assert model.Status==GRB.OPTIMAL
        installed = int(round(count.X.item()))
        assert abs(count.X.item()-installed)<1e-5 and S.X.sum()<=240+1e-6
        return S.X.copy(),installed
```

### 含储能的联合规划模型汇总

固定扰动 $P$，令 $M=\sum_i m_iS_i/S_{\rm B}$、$D=\sum_i d_iS_i/S_{\rm B}$、$K_i=k_iS_i/S_{\rm B}$。储能按 $R=r_BN$ 全额响应。将式（18″）至（22″）及网络线性化合并，得到

$$
\boxed{\begin{aligned}
\min_{S,N,q,h,z}\quad &\sum_{i=1}^{3}c_iS_i+c_BN &&\text{(18'')}\\
\text{s.t.}\quad &f_0P\le R_{\max}M &&\text{(19'')}\\
&-F_{\rm ss,max}\!\left(D+\sum_iK_i\right)\le f_0(P-r_BN)
\le F_{\rm ss,max}\!\left(D+\sum_iK_i\right) &&\text{(20'')}\\
&\widehat f_{\rm nadir}(S_1,S_2,S_3,P,N)=f_0+b+\sum_{j\in\mathcal J_L} v_jh_j^{(L)}\ge f_{\min} &&\text{(21'')}\\
&h^{(0)}=(S_1/S_{\rm B},S_2/S_{\rm B},S_3/S_{\rm B},P,N/N_{\rm sc})^\top,\\
&0\le N\le N_{\max},\quad
N_{\max}=\left\lfloor\frac{P+F_{\rm ss,max}G_{\max}/f_0}{r_B}\right\rfloor &&\text{(21''a)}\\
&q_j^{(l)}=a_j^{(l)}+\sum_{i\in\mathcal J_{l-1}}w_{ij}^{(l)}h_i^{(l-1)}
\quad(l=1,\ldots,L) &&\text{(21''b)}\\
&L_j^{(1)}=c_j+\frac{S_{\max}}{S_{\rm B}}\min(0,w_{1j}^{(1)},w_{2j}^{(1)},w_{3j}^{(1)})
+\frac{N_{\max}}{N_{\rm sc}}\min(0,w_{5j}^{(1)}),\\
&U_j^{(1)}=c_j+\frac{S_{\max}}{S_{\rm B}}\max(0,w_{1j}^{(1)},w_{2j}^{(1)},w_{3j}^{(1)})
+\frac{N_{\max}}{N_{\rm sc}}\max(0,w_{5j}^{(1)}) &&\text{(21''c)}\\
&L_j^{(l)}=a_j^{(l)}+\sum_i(w_{ij}^{(l)+}\underline h_i^{(l-1)}+w_{ij}^{(l)-}\overline h_i^{(l-1)}),\\
&U_j^{(l)}=a_j^{(l)}+\sum_i(w_{ij}^{(l)+}\overline h_i^{(l-1)}+w_{ij}^{(l)-}\underline h_i^{(l-1)})
\quad(l\ge2) &&\text{(21''d)}\\
&h_j^{(l)}\ge0,\quad h_j^{(l)}\ge q_j^{(l)},\quad
h_j^{(l)}\le U_j^{(l)}z_j^{(l)},\\
&h_j^{(l)}\le q_j^{(l)}-L_j^{(l)}(1-z_j^{(l)}),\quad
z_j^{(l)}\in\{0,1\}\quad(L_j^{(l)}<0<U_j^{(l)}) &&\text{(21''e)}\\
&h_j^{(l)}=0\ (U_j^{(l)}\le0),\qquad
h_j^{(l)}=q_j^{(l)}\ (L_j^{(l)}\ge0) &&\text{(21''f)}\\
&\widehat f_{\rm nadir}=f_0+b+\sum_{j\in\mathcal J_L}v_jh_j^{(L)} &&\text{(21''g)}\\
&\widehat f_{\rm nadir}\ge f_{\min} &&\text{(21''h)}\\
&S_i\ge0,\quad\sum_i S_i\le S_{\max},\quad N\in\mathbb Z_{\ge0}. &&\text{(22'')}
\end{aligned}}
$$

其中 $c_j=a_j^{(1)}+w_{4j}^{(1)}P$，$w^+=\max(w,0)$、$w^-=\min(w,0)$，且 $\underline h_i^{(l)}=\max(0,L_i^{(l)})$、$\overline h_i^{(l)}=\max(0,U_i^{(l)})$。式（21″c）和（21″d）的界在固定 $P$ 后预先计算，为混合整数线性规划中的常数。最低频率约束是网络预测约束，求得的方案仍须代入动态模型复核。


### 含储能的解析联合规划模型

固定扰动 $P>0$，定义 $M=\sum_i m_i S_i/S_{\rm B}$、$D=\sum_i d_i S_i/S_{\rm B}$、$K_i=k_iS_i/S_{\rm B}$。记 $I=\{i:S_i>0\}$。储能在交付时间 $T_s$ 内线性爬坡，总交付量为 $r_BN$。零初始条件下，频率偏差的拉普拉斯变换为

$$
\Delta F(s)=H(s)\left[-\frac{P}{s}+\frac{r_BN(1-e^{-sT_s})}{T_s s^2}\right],
\qquad
H(s)=\frac{f_0}{Ms+D+\displaystyle\sum_{i\in I}\frac{K_i}{1+\tau_i s}}.
$$

令 $g_1(t)=\mathcal L^{-1}\{H(s)/s\}(t)$，$g_2(t)=\mathcal L^{-1}\{H(s)/s^2\}(t)$，并令 $g_2(t)=0$ 对所有 $t<0$ 成立。频率偏差解析式为

$$
\Delta f(t)=-P g_1(t)+\frac{r_BN}{T_s}\bigl[g_2(t)-g_2(t-T_s)\bigr].
$$

对互异闭环极点 $r_j$，记 $A_j=\operatorname{Res}_{s=r_j}H(s)$，则对 $t\ge0$ 有

$$
\begin{aligned}
g_1(t)&=\sum_j A_j\frac{e^{r_jt}-1}{r_j},\\
g_2(t)&=\sum_j A_j\frac{e^{r_jt}-1-r_jt}{r_j^2}.
\end{aligned}
$$

若有重根，可直接对上式定义中的传递函数作反拉普拉斯变换。由实际频率 $f(t)=f_0+\Delta f(t)$ 得到联合规划模型

$$
\boxed{\begin{aligned}
\min_{S,N}\quad &\sum_i c_iS_i+c_BN &&\text{(18'')}\\
\text{s.t.}\quad & f_0P\le R_{\max}M &&\text{(19'')}\\
&\left|f_0(P-r_BN)\right|\le F_{\rm ss,max}\left(D+\sum_iK_i\right) &&\text{(20'')}\\
&f_0+\inf_{t\ge0}\left\{-Pg_1(t)+\frac{r_BN}{T_s}
 \bigl[g_2(t)-g_2(t-T_s)\bigr]\right\}\ge f_{\min} &&\text{(21'')}\\
&S_i\ge0,\quad \sum_i S_i\le S_{\max},\quad N\in\mathbb Z_{\ge0} &&\text{(22'')}
\end{aligned}}
$$

给定容量与储能台数后，式（21″）比较初值、交付时刻 $T_s$、两段响应的驻点及准稳态极限，取其中的最低频率。闭环极点随装机容量变化，因此该解析谷值约束一般为非线性约束。

### 同一解析模型的三种求解策略

对每个固定扰动 $P$，在同一模型中优化 $S_1,S_2,S_3\ge0$ 与整数 $N\in\{0,\ldots,N_{\max}(P)\}$，目标为 $\sum_i c_iS_i+c_BN$。储能额定功率为 $NP_B^{\max}$，保留式（19″）至（22″）中的三项频率约束与容量约束；式（21″）使用三机解析响应的谷值，不使用神经网络。

1. **逐台数枚举 + 解析 B&B（蓝）**：沿用图 11、12 的蓝线方法，先按固定台数的 LP 下界筛选，并用 SLSQP 找可行解，再对机组容量做空间分支定界。
2. **统一 MINLP + B&B（橙）**：从整个整数台数区间开始；节点暂去掉谷值约束并放松台数整数性，用 LP 求下界。只处理仍可能改善成本的区间，台数固定后继续对机组容量分支。
3. **统一 MINLP + 连续松弛热启动 + B&B（绿）**：先保留谷值约束而将 $N$ 连续化，用 SLSQP 搜索；对连续台数上下取整，分别重新优化机组容量，只用可行方案更新成本上界。若连续搜索未收敛，则试用根节点 LP 的台数作为取整起点。之后按策略 2 搜索整个整数可行域，局部松弛解不作为最优性下界。

三种策略的绝对成本间隙容差均为 0.01 百万元/年，每个工况的队列展开预算均为 2,000 个节点。计时包括初始解搜索与分支定界，不包括独立时域积分复核；空心圈表示返回方案但成本间隙未达标。扫描 $P=0.01$ 至 $1.00$，图中仅显示至少一种策略有解的范围；虚线为时域复核的频率限值。

### 解析谷值约束的全局搜索流程

固定扰动后枚举储能台数，先用线性规划筛选并以 SLSQP 寻找初始可行解。随后按以下流程验证成本最优性。

~~~text
固定储能台数 N，暂去掉动态谷值约束（21″）
保留 RoCoF、准稳态频差与容量约束，求解线性松弛
                           ↓
松弛最优成本给出该台数的下界，松弛不可行则排除该台数
                           ↓
恢复谷值约束 → SLSQP 寻找可行方案，建立成本上界
                           ↓
对各台数建立连续容量搜索区域并计算线性成本下界
                           ↓
选取当前下界最低的区域，检查成本界与动态频率包络
                           ↓
可排除则剪枝，可行则尝试更新上界，其余划分后继续搜索
                           ↓
上下界差达到成本容差 → 输出方案、成本界和求解耗时
~~~

**图 13　解析谷值约束的空间分支定界流程**

去掉式（21″）后，原问题的可行域包含于线性松弛的可行域。因此，对于固定台数或给定容量区域，线性松弛的最优目标值 $L$ 是原问题最优成本的下界。松弛解需要通过谷值约束检查，才能用其成本更新可行上界 $U$。

**算法 1　基于线性松弛的空间分支定界**

输入为固定扰动 $P$、机组与储能参数、绝对成本容差 $\varepsilon$ 和节点预算。输出为最佳可行方案、成本上界、全局下界及终止状态。

~~~text
1  枚举整数台数 N，求解删除式（21″）后的线性松弛。
2  排除松弛不可行的台数，按松弛最优成本升序处理其余台数。
3  对有机会改善成本的台数，用 SLSQP 求解含式（21″）的连续子问题。
   以通过约束检查的最低成本初始化 U。若无可行方案，令 U = +∞。
4  对各台数建立初始容量搜索区域，求取区域下界，并维护按下界排序的队列。
5  对每个待加入队列的区域 B：
      求解区域 B 的线性松弛，记最优成本为 L(B)。
      若松弛不可行或 L(B) ≥ U，则剪枝。
      若松弛解满足原约束，则用其成本更新 U。
      若某一时刻的频率上界低于安全限值，则剪枝。
      否则将 B 与其下界加入队列。
6  队列非空且 U − min L(B) > ε 时，取下界最小的区域。
   沿最宽容量区间二分，将两个子区域按步骤 5 处理，直至达到节点预算或终止条件。
7  返回最佳可行方案、剩余区域给出的全局下界及最优性间隙。
   若间隙未达到 ε，报告当前可行方案和未闭合的间隙。
~~~

算例采用浮点数计算，绝对成本容差为 0.01 百万元/年。返回的方案另用动态模型核对三项频率指标。
