# （2）EFR 调频资源的技术经济分析

## （i）资源调查

本文选择电池储能提供的增强频率响应（Enhanced Frequency Response，EFR）。电池储能通过变流器快速调整有功功率，在系统发生功率缺额时增加功率注入，减缓频率下降。

[Badesa、Teng 和 Strbac（2019） ](https://ieeexplore.ieee.org/document/8667397) 将 EFR 建模为事故后的线性爬坡响应：功率从零开始增加，在 $0.5$ s 内达到安排值。

EFR 的运行能力受储能单元功率、荷电状态、可用电量、充放电效率和响应持续时间限制。

其成本主要包括电池及变流器投资、安装、运维、电池增容以及充放电损耗。

## （ii）EFR 模型及其对频率响应的影响

### 1. EFR 容量

选择电池储能提供增强频率响应（EFR）。设系统接入多个相同储能单元，每个单元提供的 EFR 容量为

$$
R_B.
$$

若参与响应的单元数量为 $N$，总 EFR 容量为

$$
R=NR_B.
$$

#### （1）EFR 模型与单母线系统

按照论文的设定，事故发生后，EFR 功率线性上升，在交付时间内达到安排容量，随后保持不变：

$$
p_B(t)=
\begin{cases}
\dfrac{R}{T_s}t, & 0\leq t<T_s,\\[6pt]
R, & t\geq T_s.
\end{cases}
$$

将其接入第（1）问的单母线系统，并保留原调速器模型：

$$
m\dot w=p_m-p_{\ell,0}-dw+p_B,
$$

$$
\tau_g\dot p_m+p_m=-k_gw,
\qquad f=f_0w.
$$

所有功率均采用标幺值。初始状态取

$$
w(0)=0,\qquad p_m(0)=0.
$$

#### （2）分段功率的拉普拉斯变换

对 EFR 功率分段积分：

$$
P_B(s)=
\int_0^{T_s}\frac{Rt}{T_s}e^{-st}\,dt
+\int_{T_s}^{\infty}Re^{-st}\,dt.
$$

爬坡阶段的积分为

$$
\int_0^{T_s}\frac{Rt}{T_s}e^{-st}\,dt
=\frac{R}{T_ss^2}
\left[1-(1+T_ss)e^{-T_ss}\right].
$$

恒功率阶段的积分为

$$
\int_{T_s}^{\infty}Re^{-st}\,dt
=\frac{R}{s}e^{-T_ss}.
$$

合并后得到

$$
P_B(s)=\frac{R(1-e^{-T_ss})}{T_ss^2}.
$$

定义系统特征多项式：

$$
q(s)=m\tau_gs^2+(m+d\tau_g)s+d+k_g.
$$

频率偏差的拉普拉斯表达式为

$$
F(s)=
\frac{f_0(1+\tau_gs)}{q(s)}
\left[
-\frac{p_{\ell,0}}{s}
+\frac{R(1-e^{-T_ss})}{T_ss^2}
\right].
$$

#### （3）分段频率响应解析式

对于欠阻尼系统，定义

$$
\alpha=\frac{m+d\tau_g}{2m\tau_g},
\qquad
\beta=
\frac{\sqrt{4m\tau_g(d+k_g)-(m+d\tau_g)^2}}
{2m\tau_g},
$$

$$
\lambda=-\alpha+i\beta,
\qquad
A=\frac{1+\tau_g\lambda}{2im\tau_g\beta}.
$$

再定义

$$
H(t)=2\operatorname{Re}
\left[A\frac{e^{\lambda t}-1}{\lambda}\right],
$$

$$
J(t)=2\operatorname{Re}\left[
A\left(\frac{e^{\lambda t}-1}{\lambda^2}
-\frac{t}{\lambda}\right)\right].
$$

反变换后，频率偏差为

$$
f(t)=
\begin{cases}
f_0\left[-p_{\ell,0}H(t)+\dfrac{R}{T_s}J(t)\right],
& 0\leq t<T_s,\\[8pt]
f_0\left[-p_{\ell,0}H(t)
+\dfrac{R}{T_s}\bigl(J(t)-J(t-T_s)\bigr)\right],
& t\geq T_s.
\end{cases}
$$

其中，第一段对应 EFR 爬坡，第二段对应 EFR 恒功率输出。

### 2. 初始 RoCoF

由于 EFR 在事故瞬间的输出为零：

$$
p_B(0^+)=0,
$$

初始 RoCoF 为

$$
\dot f(0^+)=-\frac{f_0p_{\ell,0}}{m}.
$$

因此，在该爬坡模型下，增加储能单元数量不改变初始 RoCoF。

### 3. 准稳态频率偏差

若 EFR 持续保持输出，系统达到平衡时：

$$
f_{\mathrm{ss}}
=-\frac{f_0(p_{\ell,0}-NR_B)}{d+k_g}.
$$

当总 EFR 容量不超过负荷扰动时，增加单元数量会减小准稳态偏差的绝对值。该式以 EFR 持续输出为前提；若储能退出或恢复充电，需要重新计算退出后的频率响应。

### 4. 频率最低点及其发生时刻

#### （1）最低点发生在 EFR 爬坡阶段

若最低点发生在爬坡阶段，其时刻由以下方程确定：

$$
2\operatorname{Re}\left\{
A\left[
\left(\frac{R}{T_s\lambda}-p_{\ell,0}\right)e^{\lambda t_1}
-\frac{R}{T_s\lambda}
\right]\right\}=0,
\qquad 0<t_1<T_s.
$$

对应的频率偏差为

$$
f_1=f_0\left[
-p_{\ell,0}H(t_1)+\frac{R}{T_s}J(t_1)
\right].
$$

此时刻方程一般不能用初等函数显式表示，需数值求根。

#### （2）最低点发生在 EFR 恒功率阶段

定义

$$
z=A\left[
-p_{\ell,0}
+\frac{R}{T_s}\frac{1-e^{-\lambda T_s}}{\lambda}
\right],
\qquad
\phi=\arg z.
$$

当上式中的复数不为零时，恒功率阶段第一个局部最低点为

$$
t_2=\frac{3\pi/2-\phi+2\pi n}{\beta},
$$

$$
n=\left\lceil
\frac{\beta T_s+\phi-3\pi/2}{2\pi}
\right\rceil.
$$

对应的频率偏差为

$$
f_2=
-\frac{f_0(p_{\ell,0}-R)}{d+k_g}
-\frac{2f_0\beta|z|}{\alpha^2+\beta^2}
e^{-\alpha t_2}.
$$

最终比较爬坡阶段候选最低点、分段边界和恒功率阶段的最低点，即可确定整个响应的频率最低点及其发生时刻。

### 5. 指标比较

以 5 个储能单元为例，其 EFR 容量为

$$
N=5,\qquad R_B=0.01\ \text{p.u.}
$$

该容量为负荷扰动的 50%，是用于计算的假设值。取论文中的 EFR 交付时间：

$$
T_s=0.5\ \text{s}.
$$

#### （1） 初始 RoCoF

EFR 在事故瞬间尚未输出，因此

$$
\dot f(0^+)
=-\frac{f_0p_{\ell,0}}{m}
=-0.5000\ \text{Hz/s}.
$$

#### （2） 准稳态频率偏差

假设 EFR 持续保持输出：

$$
f_{\mathrm{ss}}
=-\frac{f_0(p_{\ell,0}-NR_B)}{d+k_g}
=-0.1190\ \text{Hz}.
$$

#### （3） 最低点及其发生时刻

代入第（1）问参数后，定义

$$
\beta=0.630476\ \text{s}^{-1}.
$$

EFR 达到安排容量后，即在 $t\geq0.5$ s 时，频率偏差为

$$
f(t)=-0.119048+e^{-0.15t}
\left[
0.056033\cos(\beta t)
-0.380028\sin(\beta t)
\right].
$$

对其求导并令导数为零：

$$
\dot f(t)=e^{-0.15t}
\left[
-0.248003\cos(\beta t)
+0.021677\sin(\beta t)
\right]=0.
$$

得到该阶段第一个局部最低点：

$$
t_{\mathrm{nadir}}=2.3532\ \text{s},
\qquad
f_{\mathrm{nadir}}=-0.3816\ \text{Hz}.
$$

检查爬坡阶段后，该点也是上述持续 EFR 模型的全局最低点。

#### （4） 指标比较

| 指标 | 不接入 EFR | 接入一个 EFR 单元 |
|---|---:|---:|
| 初始 RoCoF | −0.5000 Hz/s | −0.5000 Hz/s |
| 准稳态频率偏差 | −0.2381 Hz | −0.1190 Hz |
| 最低点频率偏差 | −0.7466 Hz | −0.3816 Hz |
| 最低点发生时刻 | 2.6170 s | 2.3532 s |

在该示例中，EFR 不改变初始 RoCoF，但使准稳态偏差和最低点偏差减小。结果假设 EFR 持续输出；退出及恢复充电后的频率变化未计入。

#### （5） 参数敏感性分析

按第（1）问参数、系统功率基准 100 MW 和 $T_s=0.5$ s 计算。每个单元 $R_B=1$ MW，即 0.01 p.u.；假设 EFR 达到安排容量后持续输出。

| 单元数 $N$ | 总 EFR 容量 | 初始 RoCoF（Hz/s） | 准稳态偏差（Hz） | 最低点偏差（Hz） | 最低点时刻（s） |
|---:|---:|---:|---:|---:|---:|
| 1 | 1 MW | −0.5000 | −0.2143 | −0.6729 | 2.5879 |
| 2 | 2 MW | −0.5000 | −0.1905 | −0.5995 | 2.5515 |
| 3 | 3 MW | −0.5000 | −0.1667 | −0.5263 | 2.5044 |
| 4 | 4 MW | −0.5000 | −0.1429 | −0.4536 | 2.4414 |
| 5 | 5 MW | −0.5000 | −0.1190 | −0.3816 | 2.3532 |

随着 $N$ 增加，最低点变浅、发生时间提前；初始 RoCoF 不变。准稳态值基于持续输出假设，未考虑 EFR 退出后的频率变化。

## （iii）参数与年化成本

参照 [Badesa、Teng 和 Strbac（2019）](https://ieeexplore.ieee.org/document/8667397) 第 IV 节英国 2030 年系统案例中的电池储能配置。将该配置作为本模型中的一个储能单元：

### 技术参数

| 参数 | 含义 | 值 | 依据 |
|---|---|---:|---|
| $P_B^{\max}$ | 单个单元额定功率 | 1 MW | 模型设定 |
| $E_B^{\max}$ | 单个单元能量容量 | 4 MWh | 按 1 MW、4 小时储能设定 |
| $T_s$ | EFR 交付时间 | 0.5 s | [Badesa 等，2019](https://doi.org/10.1109/TPWRS.2019.2905037)，第 IV 节 |
| $\eta_{\mathrm{rt}}$ | 往返效率 | 90% | Badesa 等，2019，第 IV 节 |
| $N$ | 储能单元数量 | 非负整数，待优化 | 决策变量 |

### 经济参数

| 参数 | 含义 | 值 | 依据 |
|---|---|---:|---|
| $c_E$ | 储能工程投资单价 | 1,023 元/kWh | [CESA 2024 年统计](https://www.nengyuanjie.net/article/109354.html)：四小时锂电储能 EPC 加权均价 |
| $L$ | 项目寿命 | 15 年 | 参考 [NREL 储能成本假设](https://atb.nlr.gov/electricity/2024/utility-scale_battery_storage) |
| $r$ | 实际贴现率 | 7%/年 | 参考 [Sandia 储能经济评价材料](https://www.sandia.gov/app/uploads/sites/273/2024/06/2_Byrne_Ray_SNL_ICC_Session4_12-14-2021.pdf)，第 13 页 |
| $c_{\mathrm{OM}}$ | 年固定运维及电池增容成本 | 初始投资的 2.5% | NREL 储能成本假设 |

### 年化成本计算

设每个储能单元的初始投资为 $I_B$，寿命为 $L$，贴现率为 $r$，每年固定运维成本为 $C_{\mathrm{OM}}$。储能群的年化成本为

$$
C_B(N)=N\left[
I_B\frac{r(1+r)^L}{(1+r)^L-1}
+C_{\mathrm{OM}}
\right].
$$

单个单元的初始投资为

$$
I_B=4{,}000\times1{,}023
=4.092\ \text{百万元}.
$$

按 15 年、7% 计算，资本回收系数为

$$
\mathrm{CRF}=
\frac{0.07(1+0.07)^{15}}{(1+0.07)^{15}-1}
\approx0.109795.
$$

单个单元的年化成本为

$$
C_{B,\mathrm{unit}}
=4.092(0.109795+0.025)
\approx0.5516\ \text{百万元/年}.
$$

因此，$N$ 个单元的年化成本为

$$
C_B(N)=0.5516N\quad\text{百万元/年}.
$$

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.lines import Line2D
from scipy import signal

# 系统参数：功率基准为 100 MW
f0 = 50.0                 # 额定频率，Hz
load_step = 0.10          # 负荷扰动，10 MW
m, d, k, tau = 10.0, 1.0, 20.0, 5.0

unit_response = 1.0 / 100.0  # 单个储能单元功率：1 MW
Ts = 0.5                     # EFR 交付时间，s
dt = 0.001
t = np.arange(0.0, 80.0 + dt / 2, dt)

# 净功率不平衡到标幺频率偏差的闭环传递函数
system = signal.TransferFunction(
    [tau, 1.0],
    [m * tau, m + d * tau, d + k]
)

# 分别计算负荷扰动和单个储能单元的响应
_, w_load, _ = signal.lsim(
    system, U=np.full_like(t, -load_step), T=t
)
ramp = unit_response * np.minimum(t / Ts, 1.0)
_, w_unit, _ = signal.lsim(system, U=ramp, T=t)

fig, ax = plt.subplots(figsize=(10, 6), constrained_layout=True)

print(" N  RoCoF (Hz/s)  Quasi-steady (Hz)  Nadir (Hz)  Nadir time (s)")

for N in range(6):
    frequency = f0 * (w_load + N * w_unit)
    nadir_index = np.argmin(frequency)
    qss = f0 * (-load_step + N * unit_response) / (d + k)

    label = "No storage" if N == 0 else f"N = {N}"
    line, = ax.plot(t, frequency, linewidth=1.8, label=label)

    # 标记最低点，并画出同色的准稳态水平线
    ax.plot(
        t[nadir_index], frequency[nadir_index],
        "o", color=line.get_color(), markersize=5
    )
    ax.axhline(qss, color=line.get_color(), linestyle=":", alpha=0.65)

    print(
        f"{N:2d} {-f0 * load_step / m:13.6f} {qss:18.6f} "
        f"{frequency[nadir_index]:11.6f} {t[nadir_index]:15.3f}"
    )

ax.set(
    xlabel="Time (s)",
    ylabel="Frequency deviation (Hz)",
    xlim=(0, 80)
)
ax.grid(True, alpha=0.25)

handles, _ = ax.get_legend_handles_labels()
handles.extend([
    Line2D([], [], color="black", marker="o",
           linestyle="None", label="Nadir"),
    Line2D([], [], color="black", linestyle=":",
           label="Quasi-steady state"),
])
ax.legend(handles=handles)

plt.show()
```
