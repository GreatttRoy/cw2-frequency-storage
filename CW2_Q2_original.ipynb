# （2）EFR 调频资源的技术经济分析

## （i）资源调查

本文选择电池储能提供的增强频率响应（Enhanced Frequency Response，EFR）。电池储能通过变流器快速调整有功功率，在系统发生功率缺额时增加功率注入，减缓频率下降。

[Badesa、Teng 和 Strbac（2019） ](https://ieeexplore.ieee.org/document/8667397) 将 EFR 建模为事故后的线性爬坡响应：功率从零开始增加，在 $0.5$ s 内达到安排值。

EFR 的运行能力受储能单元功率、荷电状态、可用电量、充放电效率和响应持续时间限制。

其成本主要包括电池及变流器投资、安装、运维、电池增容以及充放电损耗。

## （ii）EFR 模型及其对频率响应的影响

选择电池储能提供增强频率响应（EFR）。设系统接入多个相同储能单元，每个单元提供的 EFR 容量为

$$
R_B.
$$

若参与响应的单元数量为 $N$，总 EFR 容量为

$$
R=NR_B.
$$

### 1. EFR 模型与单母线系统

按照论文的设定，事故发生后，EFR 功率线性上升，在交付时间内达到安排容量，随后保持不变：

$$
p_B(t)=
\begin{cases}
\dfrac{R}{T_s}t, & 0\leq t<T_s,\\[6pt]
R, & t\geq T_s.
\end{cases}
$$

将其接入第（1）问的单母线系统，并保留原调速器模型：

$$
m\dot w=p_m-p_{\ell,0}-dw+p_B,
$$

$$
\tau_g\dot p_m+p_m=-k_gw,
\qquad f=f_0w.
$$

所有功率均采用标幺值。初始状态取

$$
w(0)=0,\qquad p_m(0)=0.
$$

### 2. 分段功率的拉普拉斯变换

对 EFR 功率分段积分：

$$
P_B(s)=
\int_0^{T_s}\frac{Rt}{T_s}e^{-st}\,dt
+\int_{T_s}^{\infty}Re^{-st}\,dt.
$$

爬坡阶段的积分为

$$
\int_0^{T_s}\frac{Rt}{T_s}e^{-st}\,dt
=\frac{R}{T_ss^2}
\left[1-(1+T_ss)e^{-T_ss}\right].
$$

恒功率阶段的积分为

$$
\int_{T_s}^{\infty}Re^{-st}\,dt
=\frac{R}{s}e^{-T_ss}.
$$

合并后得到

$$
P_B(s)=\frac{R(1-e^{-T_ss})}{T_ss^2}.
$$

定义系统特征多项式：

$$
q(s)=m\tau_gs^2+(m+d\tau_g)s+d+k_g.
$$

频率偏差的拉普拉斯表达式为

$$
F(s)=
\frac{f_0(1+\tau_gs)}{q(s)}
\left[
-\frac{p_{\ell,0}}{s}
+\frac{R(1-e^{-T_ss})}{T_ss^2}
\right].
$$

### 3. 分段频率响应解析式

对于欠阻尼系统，定义

$$
\alpha=\frac{m+d\tau_g}{2m\tau_g},
\qquad
\beta=
\frac{\sqrt{4m\tau_g(d+k_g)-(m+d\tau_g)^2}}
{2m\tau_g},
$$

$$
\lambda=-\alpha+i\beta,
\qquad
A=\frac{1+\tau_g\lambda}{2im\tau_g\beta}.
$$

再定义

$$
H(t)=2\operatorname{Re}
\left[A\frac{e^{\lambda t}-1}{\lambda}\right],
$$

$$
J(t)=2\operatorname{Re}\left[
A\left(\frac{e^{\lambda t}-1}{\lambda^2}
-\frac{t}{\lambda}\right)\right].
$$

反变换后，频率偏差为

$$
f(t)=
\begin{cases}
f_0\left[-p_{\ell,0}H(t)+\dfrac{R}{T_s}J(t)\right],
& 0\leq t<T_s,\\[8pt]
f_0\left[-p_{\ell,0}H(t)
+\dfrac{R}{T_s}\bigl(J(t)-J(t-T_s)\bigr)\right],
& t\geq T_s.
\end{cases}
$$

其中，第一段对应 EFR 爬坡，第二段对应 EFR 恒功率输出。

### 4. 初始 RoCoF

由于 EFR 在事故瞬间的输出为零：

$$
p_B(0^+)=0,
$$

初始 RoCoF 为

$$
\dot f(0^+)=-\frac{f_0p_{\ell,0}}{m}.
$$

因此，在该爬坡模型下，增加储能单元数量不改变初始 RoCoF。

### 5. 准稳态频率偏差

若 EFR 持续保持输出，系统达到平衡时：

$$
f_{\mathrm{ss}}
=-\frac{f_0(p_{\ell,0}-NR_B)}{d+k_g}.
$$

当总 EFR 容量不超过负荷扰动时，增加单元数量会减小准稳态偏差的绝对值。该式以 EFR 持续输出为前提；若储能退出或恢复充电，需要重新计算退出后的频率响应。

### 6. 频率最低点及其发生时刻

#### （1）最低点发生在 EFR 爬坡阶段

若最低点发生在爬坡阶段，其时刻由以下方程确定：

$$
2\operatorname{Re}\left\{
A\left[
\left(\frac{R}{T_s\lambda}-p_{\ell,0}\right)e^{\lambda t_1}
-\frac{R}{T_s\lambda}
\right]\right\}=0,
\qquad 0<t_1<T_s.
$$

对应的频率偏差为

$$
f_1=f_0\left[
-p_{\ell,0}H(t_1)+\frac{R}{T_s}J(t_1)
\right].
$$

此时刻方程一般不能用初等函数显式表示，需数值求根。

#### （2）最低点发生在 EFR 恒功率阶段

定义

$$
z=A\left[
-p_{\ell,0}
+\frac{R}{T_s}\frac{1-e^{-\lambda T_s}}{\lambda}
\right],
\qquad
\phi=\arg z.
$$

当上式中的复数不为零时，恒功率阶段第一个局部最低点为

$$
t_2=\frac{3\pi/2-\phi+2\pi n}{\beta},
$$

$$
n=\left\lceil
\frac{\beta T_s+\phi-3\pi/2}{2\pi}
\right\rceil.
$$

对应的频率偏差为

$$
f_2=
-\frac{f_0(p_{\ell,0}-R)}{d+k_g}
-\frac{2f_0\beta|z|}{\alpha^2+\beta^2}
e^{-\alpha t_2}.
$$

最终比较爬坡阶段候选最低点、分段边界和恒功率阶段的最低点，即可确定整个响应的频率最低点及其发生时刻。增加储能单元数量会改变上述最低点公式中的总 EFR 容量；最低点时刻的变化方向需结合系统参数计算。

## （iii）参数与年化成本

选用 [Badesa、Teng 和 Strbac（2019）](https://ieeexplore.ieee.org/document/8667397) 第 IV 节英国 2030 年系统案例中的电池储能配置。将该配置作为本模型中的一个储能单元：

**技术参数：**

| 参数 | 含义 | 值 | 依据 |
|---|---|---:|---|
| $P_B^{\max}$ | 单个储能单元额定功率 | 200 MW | 论文第 IV 节 |
| $E_B^{\max}$ | 单个储能单元能量容量 | 1,000 MWh | 论文第 IV 节 |
| $T_s$ | EFR 交付时间 | 0.5 s | 论文第 IV 节 |
| $\eta_{\mathrm{rt}}$ | 储能往返效率 | 90% | 论文第 IV 节 |
| $R_B$ | 单个单元安排的 EFR 容量 | 200 MW | 按事故前电池静止、额定功率为 200 MW 设定 |
| $N$ | 相同储能单元数量 | 非负整数，待优化 | 模型决策变量 |
$c_E$ | 储能工程投资单价 | 1,023 元/kWh | [CESA 2024 年招标统计](https://finance.sina.cn/2025-01-15/detail-ineeznmf7350490.d.html?vt=4)：四小时锂电储能 EPC 加权均价 |
| $L$ | 项目寿命 | 15 年 | 参考 [NREL 储能成本假设](https://atb.nlr.gov/electricity/2024/utility-scale_battery_storage) |
| $r$ | 贴现率 | 7%/年 | 计算假设；应根据项目融资条件调整 |
| $c_{\mathrm{OM}}$ | 年固定运维及电池增容成本 | 初始投资的 2.5% | NREL 储能成本假设 |

**年化成本：**

设每个储能单元的初始投资为 $I_B$，寿命为 $L$，贴现率为 $r$，每年固定运维成本为 $C_{\mathrm{OM}}$。储能群的年化成本为

$$
C_B(N)=N\left[
I_B\frac{r(1+r)^L}{(1+r)^L-1}
+C_{\mathrm{OM}}
\right].
$$
