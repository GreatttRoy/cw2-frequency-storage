## 由式（1）–（3）建立控制框图
令 $P=0.10$ 为负荷阶跃幅值。沿用式（1）的 $\omega$ 表示标幺频率偏差，则 $\Delta f=f_0\omega$，实际频率为 $f_0+\Delta f$。初始平衡给出 $\omega(0^-)=p_m(0^-)=0$。

```text
                                 ┌──────[ −k/(1+τs) ]◄────┐
                                 ▼ +                      │
Pℓ(s) ────────────────────(−)──► Σ ───► [ 1/(ms+d) ] ────► Ω(s) ──► [ f₀ ] ──► ΔF(s)
```

式（1）移项得 $m\dot\omega+d\omega=p_m-p_l$。取拉普拉斯变换得 $(ms+d)\Omega=P_m-P_l$。用式（3）代入 $P_m=-k\Omega/(1+\tau s)$，两边同乘 $1+\tau s$，得到
$$
[(ms+d)(1+\tau s)+k]\Omega=-(1+\tau s)P_l. \qquad (4)
$$
将乘积展开并代入式（2）的 $P_l=P/s$，可得
$$
\Delta F(s)=-\frac{f_0P(1+\tau s)}{s[m\tau s^2+(m+d\tau)s+d+k]}. \qquad (5)
$$

### (i) 初始频率变化率
由式（3）得 $\tau\dot p_m+p_m=-k\omega$。有限幅度阶跃下，惯量和调速器使 $\omega(0^+)=p_m(0^+)=0$。将式（2）的 $p_l(0^+)=P$ 代入式（1）：
$$
m\dot\omega(0^+)=0-P-d\cdot0=-P. \qquad (6)
$$
因此
$$
\dot{\Delta f}(0^+)=f_0\dot\omega(0^+)=-\frac{f_0P}{m}=-\frac{50\times0.10}{10}=-0.500000\ \mathrm{Hz/s}. \qquad (7)
$$

### (ii) 准稳态频率偏差
由调速器传递函数 $P_m/\Omega=-k/(1+\tau s)$，两边交叉相乘并转换到时域：
$$
\tau\dot p_m+p_m=-k\omega. \qquad (8)
$$
稳态时 $\dot p_m=0$，负荷扰动 $p_{l,\mathrm{ss}}=P$，因此
$$
p_{m,\mathrm{ss}}=-k\omega_{\mathrm{ss}}. \qquad (9)
$$
在式（1）中令 $\dot\omega=0$，并代入上述稳态关系：
$$
0=-k\omega_{\mathrm{ss}}-P-d\omega_{\mathrm{ss}}
\quad\Rightarrow\quad (d+k)\omega_{\mathrm{ss}}=-P. \qquad (10)
$$
结合 $\Delta f_{\mathrm{ss}}=f_0\omega_{\mathrm{ss}}$，得到
$$
\Delta f_{\mathrm{ss}}=-\frac{f_0P}{d+k}=-\frac{5}{21}=-0.238095\ \mathrm{Hz}. \qquad (11)
$$
对应实际频率为 $49.761905\ \mathrm{Hz}$。

### (iii) 由拉普拉斯反变换求最低点
对频率偏差的拉普拉斯表达式（5），先消去阶跃项的 $1/s$，再将所得式的分子和闭环分母同时除以 $m\tau$。记 $a=(m+d\tau)/(2m\tau)$，$b=\sqrt{(d+k)/(m\tau)-a^2}$，$C=f_0P/m$。本算例 $b>0$，故二次项可配方为 $(s+a)^2+b^2$。由上面的频率偏差表达式，两边乘以 $s$，再将分子和分母同时除以 $m\tau$，得到
$$
s\Delta F(s)=-\frac{f_0P}{m}
\frac{s+1/\tau}{s^2+\dfrac{m+d\tau}{m\tau}s+\dfrac{d+k}{m\tau}}. \qquad (11a)
$$
由于初始频率偏差 $\Delta f(0^+)=0$，有 $\mathcal L\{\dot{\Delta f}(t)\}=s\Delta F(s)$。根据 $a$、$b$ 的定义，分母配方为
$$
s^2+\frac{m+d\tau}{m\tau}s+\frac{d+k}{m\tau}=(s+a)^2+b^2. \qquad (11b)
$$
将分子拆成 $s+1/\tau=(s+a)+(1/\tau-a)$，并代入 $C=f_0P/m$，得到
$$
s\Delta F(s)=-C\frac{s+1/\tau}{(s+a)^2+b^2}
=-C\left[\frac{s+a}{(s+a)^2+b^2}+\frac{1/\tau-a}{b}\frac{b}{(s+a)^2+b^2}\right]. \qquad (12)
$$
令 $h=(1/\tau-a)/b$，逐项反变换得到
$$
\dot{\Delta f}(t)=-Ce^{-at}[\cos(bt)+h\sin(bt)]. \qquad (13)
$$
令中括号为零，取第一个正根 $bt=\operatorname{atan2}(b,a-1/\tau)$。导数在此由负转正，随后的振荡按 $e^{-at}$ 衰减，故首次谷值即最低点。代入 $a=0.15$、$b=\sqrt{0.42-0.15^2}$：
$$
t_{\rm nadir}=\frac{\operatorname{atan2}(b,-0.05)}{b}=2.616968\ \mathrm{s}. \qquad (14)
$$
从 $0$ 到 $t$ 积分导数，并用 $\Delta f(0)=0$ 确定积分常数：
$$
\Delta f(t)=-C\int_0^t e^{-au}[\cos(bu)+h\sin(bu)]\,du
=-\frac{C}{a^2+b^2}\left[a+hb+e^{-at}\big(-(a+hb)\cos(bt)+(b-ha)\sin(bt)\big)\right]. \qquad (15)
$$
因 $a+hb=1/\tau$，代入基准参数及上述时刻得
$$
\Delta f(t_{\rm nadir})=-\frac{0.5}{0.42}\left[0.2+e^{-0.15t}\big(-0.2\cos(bt)+(b-0.15h)\sin(bt)\big)\right]_{t=t_{\rm nadir}}
=-0.746571\ \mathrm{Hz}. \qquad (16)
$$
对应实际最低频率为 $49.253429\ \mathrm{Hz}$。

将公式（14）的首个极小值时刻代入式（15），并利用 $b^2+(a-1/\tau)^2=k/(m\tau)$，可将最低频率偏差写成关于模型参数的数学形式：
$$
\Delta f_{\mathrm{nadir}}
=-\frac{f_0P}{d+k}\left[1+\sqrt{\frac{k\tau}{m}}\exp\left(-\frac{a}{b}\operatorname{atan2}\left(b,a-\frac{1}{\tau}\right)\right)\right],
\quad a=\frac{m+d\tau}{2m\tau},\quad b=\sqrt{\frac{d+k}{m\tau}-a^2}. \qquad (17)
$$
式（17）适用于 $m,\tau,k,P>0$、$d\geq 0$ 且 $b>0$ 的欠阻尼情形。若需要实际最低频率，则 $f_{\mathrm{nadir}}=f_0+\Delta f_{\mathrm{nadir}}$。

### (iv) 时域仿真与误差
用 python-control 的负反馈连接摆动环节和调速器，再在负荷输入端乘以负号。表中初始斜率采用三点前向差分，稳态值取最后一秒的均值，最低点取离散曲线的最小样本。

````python
# 静默当前内核反复报告的矩阵乘法和内嵌图表提示。
import warnings
warnings.filterwarnings("ignore", message=r"(divide by zero|overflow|invalid value) encountered in matmul$", category=RuntimeWarning)
warnings.filterwarnings("ignore", message=r"Consider using IPython.display.IFrame instead", category=UserWarning)

import control as ct  # python-control：传递函数、反馈连接和阶跃响应
import slycot  # 可选后端，用于确认当前环境已安装
import sys
print(f"Python: {sys.executable} | control: {ct.__version__} | slycot: {slycot.__version__}")
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display

base = dict(m=10., d=1., k=20., tau=5.)
f0, P = 50., 0.10

def simulate(m, d, k, tau, dt=0.001):
    swing = ct.tf([1], [m, d])
    governor = ct.tf([k], [tau, 1])
    system = -f0 * P * ct.feedback(swing, governor)
    end = 12 / min(-ct.poles(system).real)
    t = np.arange(int(np.ceil(end / dt)) + 1) * dt
    t, y = ct.step_response(system, T=t)
    y = np.asarray(y)
    i = y.argmin()
    metrics = [(-3*y[0]+4*y[1]-y[2])/(2*dt), y[-round(1/dt):].mean(), y[i], t[i]]
    return t, y, metrics

a = 0.15
b = np.sqrt(0.42-a*a)
h = (0.2-a)/b
def analytical(t):
    return -0.5/0.42*(0.2+np.exp(-a*t)*(-0.2*np.cos(b*t)+(b-a*h)*np.sin(b*t)))

tn = np.arctan2(b, -0.05)/b
exact = np.array([-0.5, -5/21, analytical(tn), tn])
t, y, measured = simulate(**base)
labels = ['RoCoF (Hz/s)', '准稳态偏差 (Hz)', '最低点偏差 (Hz)', '最低点时刻 (s)']
display(pd.DataFrame({'仿真值': measured, '解析值': exact, '绝对误差': abs(measured-exact)}, index=labels).style.format('{:.9f}'))
_, _, finer = simulate(**base, dt=0.0005)
assert abs(finer[3]-tn) <= 0.0005 and abs(measured[3]-tn) <= 0.001
fig, ax = plt.subplots()
ax.plot(t, y, label='python-control')
ax.plot(t, analytical(t), '--', label='Analytical')
ax.plot(tn, analytical(tn), 'ro', label='Nadir')
ax.axhline(exact[1], color='gray', ls=':', label='Quasi-steady state')
ax.set(xlabel='Time (s)', ylabel='Frequency deviation (Hz)')
ax.legend()
plt.show()
````

## 以机组容量为决策变量的模型

令 $S=(S_1,S_2,S_3)$，其中 $S_i$ 为第 $i$ 台机组的安装容量（MW）。优化只选择这三个容量。负荷阶跃幅值 $P$ 固定；惯量、阻尼、机械功率响应及频率指标均由 $S$ 确定。

### 目标函数与约束

以下求和均遍历参数表中的三台机组。按 CW2.pdf 第三问，式（20′）采用阻尼与调速增益之和，式（21′）采用题设的最低点限值。三个调速器各自保留其时间常数。

**年化成本最小化**（百万元/年）

$$
\min_{S_1,S_2,S_3}\ \sum_i c_i S_i . \tag{18′}
$$

**初始 RoCoF 约束。** 由 $|\dot{\Delta f}(0^+;S)|=f_0PS_{\rm B}/\sum_i m_iS_i$，可将变化率上限写成容量加权惯量的下限。因右端为正，该约束同时排除了全零容量。

$$
\sum_i m_i S_i \ge \frac{f_0 P S_{\rm B}}{R_{\max}}. \tag{19′}
$$

**准稳态频差约束**（Hz）

$$
|\Delta f_{\rm ss}(S)|=
\frac{f_0 P S_{\rm B}}{\sum_i (d_i+k_i)S_i}\le F_{\rm ss,max}. \tag{20′}
$$

**最低点频差约束**（Hz）。对正负荷阶跃，取频率下降量随时间的最大值。将各机组响应代入摆动方程，得到

$$
|\Delta f_{\rm nadir}(S)|=
\sup_{t\ge 0}\mathcal L^{-1}\!\left\{
\frac{f_0 P S_{\rm B}}
{s\displaystyle\sum_i S_i\!\left(m_i s+d_i+\frac{k_i}{1+\tau_i s}\right)}
\right\}(t)\le F_{\rm nad,max}. \tag{21′}
$$

**式（21′）的完整时域形式。** 三台机组均投入且闭环极点互异时，取 $n=4$。将反拉普拉斯变换逐项算出并代回最低点约束，得到同一条式（21′）：

$$
\boxed{
|\Delta f_{\rm nadir}(S)|
=f_0 P S_{\rm B}\max_{t\in\mathcal T(S)}
\left[
\frac{1}{\displaystyle\sum_{i=1}^{3}(d_i+k_i)S_i}
+
\sum_{j=1}^{n}
\frac{\displaystyle\prod_{i\in I}(1+\tau_i r_j)\,e^{r_jt}}
{\displaystyle r_j
 \left(\sum_{i=1}^{3}m_i S_i\right)
 \left(\prod_{i\in I}\tau_i\right)
 \prod_{\substack{1\le\ell\le n\\\ell\ne j}}(r_j-r_\ell)}
\right]\le F_{\rm nad,max}.
}
$$

这里 $I=\{i:S_i>0\}$。部分机组容量为零时，只取投入的调速器，令 $n=|I|+1$。极点 $r_1,\ldots,r_n$ 均由容量决定，是下列多项式的根：

$$
\left[\sum_{i\in I} S_i(m_i r+d_i)\right]\prod_{h\in I}(1+\tau_h r)
+\sum_{i\in I}k_i S_i\prod_{h\in I\setminus\{i\}}(1+\tau_h r)=0 .
$$

方框内第一项来自阶跃在零点的极点，其余各项来自闭环极点。对方括号中的时域表达式求导，每项指数前的 $r_j$ 与分母中的 $r_j$ 抵消。消去公共因子后，得到需要比较的全部候选时刻：

$$
\mathcal T(S)=\{0,+\infty\}\ \cup\!
\left\{t>0:\ 
\sum_{j=1}^{n}
\frac{\displaystyle\prod_{i\in I}(1+\tau_i r_j)}
     {\displaystyle\prod_{\substack{1\le\ell\le n\\\ell\ne j}}(r_j-r_\ell)}
 e^{r_jt}=0\right\}.
$$

$+\infty$ 表示取稳态极限。共轭极点的两项相加为实数。若极点重合，方框中相应指数项之和按极点合并时的连续极限取值，再对所得时域函数求驻点。三种时间常数各异时，驻点条件一般是多个指数项组成的超越方程；给定容量后须对时间求根，因此没有第一问单一调速器模型中通用的反正切峰值时间公式。

**容量约束**

$$
S_i\ge0\quad(\forall i),\qquad \sum_i S_i\le S_{\max}. \tag{22′}
$$

### 式（21′）方框内公式的推导

**1. 从摆动方程得到频率下降量的拉普拉斯变换。** 令 $y(t;S)=-\Delta f(t;S)$，记其拉普拉斯变换为 $Y(s;S)$。零初始条件下，将调速器的机械功率增量代入式（1），并在两边同乘系统基准 $S_{\rm B}$，得到

$$
\left[\sum_{i\in I} S_i(m_i s+d_i)
      +\sum_{i\in I}\frac{k_i S_i}{1+\tau_i s}\right]\Omega(s)
=-S_{\rm B}\frac{P}{s}.
$$

由 $Y=-f_0\Omega$，令 $C=f_0PS_{\rm B}$，则

$$
Y(s;S)=\frac{C}{s\displaystyle\sum_{i\in I}S_i
\left(m_i s+d_i+\frac{k_i}{1+\tau_i s}\right)}.
$$

**2. 通分并分解闭环分母。** 令 $Q(s)=\prod_{i\in I}(1+\tau_i s)$。上式分子、分母同乘 $Q(s)$，得到

$$
Y(s;S)=\frac{C Q(s)}{s B(s)},\qquad
B(s)=\left[\sum_{i\in I}S_i(m_i s+d_i)\right]Q(s)
+\sum_{i\in I} k_iS_i\prod_{h\in I\setminus\{i\}}(1+\tau_hs).
$$

$B(s)$ 的最高次项来自 $s(\sum_i m_i S_i)Q(s)$。记最高次系数为 $L$，次数为 $n=|I|+1$，则

$$
L=\left(\sum_i m_i S_i\right)\prod_{i\in I}\tau_i,\qquad
B(s)=L\prod_{j=1}^n(s-r_j).
$$

**3. 求各极点的部分分式系数。** 在闭环极点互异时，写成 $Y(s;S)=C_0/s+\sum_{j=1}^n C_j/(s-r_j)$。由 $Q(0)=1$ 和 $B(0)=\sum_i(d_i+k_i)S_i$，得到

$$
C_0=\lim_{s\to0}sY(s;S)
=\frac{C}{\sum_i(d_i+k_i)S_i}.
$$

对极点 $r_j$，用 $s-r_j$ 乘 $Y$，约去 $B(s)$ 中对应的因子，再取 $s\to r_j$，得到

$$
C_j=\lim_{s\to r_j}(s-r_j)Y(s;S)
=\frac{C\displaystyle\prod_{i\in I}(1+\tau_i r_j)}
{r_j L\displaystyle\prod_{\ell\ne j}(r_j-r_\ell)} .
$$

**4. 逐项反变换并代回最低点约束。** 由 $1/s$ 和 $1/(s-r_j)$ 分别对应时域函数 $1$ 和 $e^{r_jt}$，将 $C_0,C_j,L$ 代回，得到

$$
y(t;S)=f_0PS_{\rm B}\left[
\frac{1}{\sum_i(d_i+k_i)S_i}
+\sum_{j=1}^n
\frac{\prod_{i\in I}(1+\tau_i r_j)\,e^{r_jt}}
{r_j(\sum_i m_iS_i)(\prod_{i\in I}\tau_i)
\prod_{\ell\ne j}(r_j-r_\ell)}
\right].
$$

对该式求导得到 $\dot y(t;S)=\sum_j r_j C_j e^{r_jt}$。令其为零，消去公共系数 $C/L$，即得式（21′）下方定义候选时刻集合 $\mathcal T(S)$ 的驻点方程。将上式代入 $|\Delta f_{\rm nadir}(S)|=\max_{t\in\mathcal T(S)}y(t;S)\le F_{\rm nad,max}$，即得方框内的不等式。

## 第三问补充推导：由机组容量得到频率安全约束

式（18）已将决策变量改为机组容量。以下按 CW2.pdf 第 2 页给出的模型校核。取 $S_i$ 的数值单位为 MW，系统有功功率基准按题给公式取 100 MW。安全校核固定 $P=0.10$ p.u.，故式（18）的优化变量为 $S_1,S_2,S_3$。采样时可以改变扰动幅值。由式（22）及题设，各容量非负且总和不超过 240 MW，计算频率响应时须有正惯量。

### 式（23）、（24）可以直接代入摆动方程

各机组处于同一母线，具有相同的标幺频率偏差。将各机组的惯量项和阻尼项按容量换算到系统基准后分别相加，得到式（23）、（24）。写开为

$$
M(S)=m=0.14S_1+0.08S_2+0.04S_3,\qquad
D(S)=d=0.008S_1+0.011S_2+0.015S_3. \tag{26}
$$

这两个量可以直接替换式（1）中的 $m,d$。式（5）还包含调速器模型，沿用其二阶分母需要另行检查式（25）。

### 将式（25）代入式（5）的推导

按题设，式（25）左端应为机械功率增量 $P_m(s)$，右端带负反馈符号，且每项增益均须除以 100 MW。令 $K_i=S_i k_i/100$，则

$$
P_m(s)=-\Omega(s)\,G_S(s),\qquad
G_S(s)=\frac{0.15S_1}{1+7s}
      +\frac{0.23S_2}{1+3s}
      +\frac{0.30S_3}{1+0.8s}. \tag{27}
$$

回到式（4）之前的摆动方程，代入式（27），再将含 $\Omega$ 的项移到左边：

$$
(Ms+D)\Omega=-G_S(s)\Omega-P_l
\quad\Longrightarrow\quad
[Ms+D+G_S(s)]\Omega=-P_l. \tag{28}
$$

令 $Q(s)=(1+7s)(1+3s)(1+0.8s)$。两边同乘 $Q(s)$，再代入 $P_l=P/s$ 和 $\Delta F=f_0\Omega$，得到第三问中与式（5）对应的频率响应：

$$
\Delta F(s;S)=-\frac{f_0P Q(s)}{sH(s;S)}, \tag{29}
$$

$$
\begin{aligned}
H(s;S)={}&[M(S)s+D(S)]Q(s)\\
 &+0.15S_1(1+3s)(1+0.8s)\\
 &+0.23S_2(1+7s)(1+0.8s)\\
 &+0.30S_3(1+7s)(1+3s).
\end{aligned} \tag{30}
$$

三台机组均投入时，$H$ 为四次多项式。若所有投入机组具有相同时间常数，式（27）可合并为一个一阶环节，式（29）便化为式（5），其中 $k=\sum_i K_i$。本题三个时间常数不同，因此不能把总增益和某个平均时间常数直接代入式（17）求谷值。

### 式（19）至（21）以容量为决策变量的形式

**式（19）初始 RoCoF。** 初始平衡时，频差及各调速器输出均为零。由式（1），$M(S)\dot\omega(0^+)=-P$。乘以 $f_0$ 并取绝对值，得到

$$
|\dot{\Delta f}(0^+;S)|=
\frac{5}{0.14S_1+0.08S_2+0.04S_3}
\le 0.50\ \mathrm{Hz/s}. \tag{31}
$$

**式（20）准稳态频差。** 依式（5）或稳态功率平衡，Word 式（20）分母的 $a$ 应改为阻尼 $d$。式（27）的直流增益为 $G_S(0)=0.15S_1+0.23S_2+0.30S_3$。在稳态，$[D(S)+G_S(0)]\omega_{\rm ss}=-P$，故

$$
|\Delta f_{\rm ss}(S)|=
\frac{5}{0.158S_1+0.241S_2+0.315S_3}
\le 0.30\ \mathrm{Hz}. \tag{32}
$$

**式（21）最低点频差。** CW2.pdf 规定限值为 $0.50$ Hz，Word 式（21）写作 $0.05$。令下式取上确界，是因为负荷增加时最低频率对应 $-\Delta f(t;S)$ 的最大值。由式（28）直接得到只含三个容量决策变量的约束：

$$
|\Delta f_{\rm nadir}(S)|=
\sup_{t\ge0}\mathcal L^{-1}\!\left\{
\frac{5}{s\left[
(0.14S_1+0.08S_2+0.04S_3)s+
0.008S_1+0.011S_2+0.015S_3+
\dfrac{0.15S_1}{1+7s}+
\dfrac{0.23S_2}{1+3s}+
\dfrac{0.30S_3}{1+0.8s}
\right]}
\right\}(t)
\le 0.50\ \mathrm{Hz}. \tag{33}
$$

给定一组容量后，还可直接从式（30）求值。若 $H$ 的四个根 $r_j$ 互异，令 $A_j=5Q(r_j)/[r_j H'(r_j;S)]$，其中导数对 $s$ 求取。对式（29）作部分分式展开，得到时域频差的相反数

$$
-\Delta f(t;S)=\frac{5}{H(0;S)}+
\sum_{j=1}^{4}A_j e^{r_jt}. \tag{34}
$$

令 $\sum_j r_j A_j e^{r_jt}=0$，求出正时间驻点，将式（34）在这些时刻的值与初值零及稳态值 $5/H(0;S)$ 比较，取最大值即为式（33）左端。若出现重根，可直接使用式（33）。由此，前两项约束可分别化为容量的线性不等式，最低点约束则保留三个调速器各自的响应时间。


## 动态仿真样本与安全标签

三个容量均按 1 MW 间隔从 1 取到 100 MW，总量至多 240 MW。负荷阶跃取 0.01 至 1.00 p.u.，步长 0.01。每个阶跃对应 962,180 条记录。以下单元可从本节开始依次运行。

````python
# 静默当前内核反复报告的矩阵乘法和内嵌图表提示。
import warnings
warnings.filterwarnings("ignore", message=r"(divide by zero|overflow|invalid value) encountered in matmul$", category=RuntimeWarning)
warnings.filterwarnings("ignore", message=r"Consider using IPython.display.IFrame instead", category=UserWarning)

import numpy as np
import pandas as pd
import pyarrow as pa
import pyarrow.parquet as pq
from pathlib import Path
from tqdm import tqdm as Tqdm
from IPython.display import display, HTML
from html import escape
from io import StringIO
import control as ct

m_i = np.array([14., 8., 4.])
d_i = np.array([0.8, 1.1, 1.5])
k_i = np.array([15., 23., 30.])
tau_i = np.array([7., 3., 0.8])
threshold = np.array([0.50, 0.30, 0.50])
grid = np.indices((100, 100, 100), dtype=np.int16).reshape(3, -1).T + 1
capacity = grid[grid.sum(axis=1) <= 240]
factor = np.gcd.reduce(capacity, axis=1)
ratio, index = np.unique(capacity // factor[:, None], axis=0, return_inverse=True)
index = index.ravel()
assert len(capacity) == 962180
directory = Path("cw2_samples")
directory.mkdir(exist_ok=True)


class NotebookProgress(Tqdm):
    """在同一个 Jupyter 输出区域更新进度，避免终端回车符产生多行。"""
    def __init__(self, *args, **kwargs):
        self._display_handle = None
        kwargs.setdefault("file", StringIO())
        kwargs.setdefault("mininterval", 0.2)
        super().__init__(*args, **kwargs)

    def display(self, msg=None, pos=None):
        text = str(self) if msg is None else str(msg)
        content = HTML("<pre style=\"margin:0;white-space:pre-wrap\">" + escape(text) + "</pre>")
        if self._display_handle is None:
            self._display_handle = display(content, display_id=True)
        else:
            self._display_handle.update(content)
````

### 参考阶跃的时域响应

每个最简整数容量配比单独求解四状态线性微分方程。以下以状态矩阵的模态展开计算时间响应，并在下降量的每个候选峰附近二分求导数零点。调速器各自保留其时间常数。同一配比下，容量同时乘以 $g$，扰动取 $P$ 时，参考频差响应乘以 $P/g$，峰值时间相同。

````python
def response_batch(S):
    """1 p.u. 阶跃，输出正的频率下降量。状态为频率和三个调速器功率增量。"""
    M = (S * m_i).sum(axis=1) / 100
    D = (S * d_i).sum(axis=1) / 100
    K = S * k_i / 100
    A = np.zeros((len(S), 4, 4))
    A[:, 0, 0] = -D / M
    A[:, 0, 1:] = 1 / M[:, None]
    A[:, 1:, 0] = -K / tau_i
    A[:, [1, 2, 3], [1, 2, 3]] = -1 / tau_i
    poles, vectors = np.linalg.eig(A)
    steady = 50 / (D + K.sum(axis=1))
    equilibrium = np.column_stack((-steady / 50, K * steady[:, None] / 50))
    weights = np.linalg.solve(vectors, equilibrium[..., None])[..., 0]
    modes = 50 * vectors[:, 0, :] * weights
    assert np.all(poles.real < 0)
    step = min(0.1, 0.19 / max(1., np.max(abs(poles.imag))))
    assert np.max(abs(poles.imag)) * step < 0.2
    advance = np.exp(poles * step)
    term = modes.copy()
    best = steady.copy()
    t_best = np.full(len(S), np.inf)
    derivative = (term * poles).sum(axis=1).real
    t = 0.
    while True:
        term *= advance
        next_derivative = (term * poles).sum(axis=1).real
        peak = (derivative > 0) & (next_derivative <= 0)
        if peak.any():
            ii = np.flatnonzero(peak)
            lo = np.full(len(ii), t)
            hi = lo + step
            for _ in range(23):
                mid = (lo + hi) / 2
                rising = ((modes[ii] * poles[ii]) * np.exp(poles[ii] * mid[:, None])).sum(axis=1).real > 0
                lo = np.where(rising, mid, lo)
                hi = np.where(rising, hi, mid)
            peak_time = (lo + hi) / 2
            value = steady[ii] + (modes[ii] * np.exp(poles[ii] * peak_time[:, None])).sum(axis=1).real
            better = value > best[ii]
            best[ii[better]] = value[better]
            t_best[ii[better]] = peak_time[better]
        derivative = next_derivative
        t += step
        # 稳定模态包络限制未来可能的极值，避免截断较晚的谷值。
        if np.all(steady + abs(term).sum(axis=1) <= best + 1e-12):
            break
    return best, t_best

reference = np.empty((len(ratio), 2))
with NotebookProgress(total=len(ratio), desc="动态仿真", unit="配比") as bar:
    for start in range(0, len(ratio), 4096):
        batch = ratio[start:start + 4096]
        reference[start:start + len(batch)] = np.column_stack(response_batch(batch))
        bar.update(len(batch))
nadir_unit = reference[index, 0] / factor
t_nadir = reference[index, 1]
````

### 生成标签与 Parquet 样本

每个分区对应一个阶跃。三项指标保存带符号值，标签按绝对值判定。严格大于 0.50 Hz/s、0.30 Hz、0.50 Hz 才记为对应越限，等于限值合格。下表显示各扰动的越限样本数。

````python
# 整数交叉相乘处理两个解析指标的阈值相等情况。
inertia = capacity @ m_i.astype(int)
response_gain10 = capacity @ np.rint(10 * (d_i + k_i)).astype(int)
names = ["rocof_hz_s", "steady_deviation_hz", "nadir_deviation_hz"]
flags = ["rocof_violation", "steady_violation", "nadir_violation"]
counts = []
with NotebookProgress(total=100, desc="样本生成与保存", unit="分区") as bar:
    for level in range(1, 101):
        P = level / 100
        values = [-50 * level / inertia, -500 * level / response_gain10, -P * nadir_unit]
        violations = [100 * level > inertia, 5000 * level > 3 * response_gain10,
                      -values[2] > threshold[2]]
        any_bad = np.logical_or.reduce(violations)
        data = {f"S{i+1}_MW": capacity[:, i].astype(np.uint8) for i in range(3)}
        data.update(P_pu=np.full(len(capacity), P), t_nadir_s=t_nadir)
        data.update(zip(names, values))
        data.update(zip(flags, violations))
        data["any_violation"] = any_bad
        dst = directory / f"step_{level:03d}.parquet"
        tmp = dst.with_suffix(".tmp")
        pq.write_table(pa.table(data), tmp, compression="zstd", compression_level=4,
                       use_dictionary=["P_pu"], use_byte_stream_split=names + ["t_nadir_s"])
        assert pq.read_metadata(tmp).num_rows == len(capacity)
        tmp.replace(dst)
        counts.append([*(int(x.sum()) for x in violations), int(any_bad.sum())])
        bar.set_postfix(P=P, rows=level * len(capacity), refresh=False)
        bar.update(1)
counts = np.array(counts)
assert len(capacity) * len(counts) == 96_218_000
print("表 1：各负荷阶跃幅值下，三项指标及任一指标越限的样本数量。")
display(pd.DataFrame(counts, columns=flags + ["any_violation"], index=np.arange(1,101)/100).rename_axis("P_pu"))
print("表 2：P = 0.10 p.u. 样本分区的前 5 行，展示容量、频率指标、最低点时间及越限标签。")
display(pq.ParquetFile(directory / "step_010.parquet").read_row_group(0).slice(0,5).to_pandas())
````

### 独立仿真校验与静态平衡

装机容量描述额定能力。扰动前的负荷和发电已处于平衡，当前增量模型无需新增静态调度等式。用末端状态检查 $\sum_i p_{m,i}-P-D\omega=0$，并用 python-control 直接阶跃仿真核对样本。

````python
# 直接在原容量及真实扰动下运行 python-control，覆盖边界和等比例工况。
P_check = 0.10
sample = pq.read_table(directory / "step_010.parquet").to_pandas()
security_ratio = np.max(np.abs(sample[names].to_numpy()) / threshold, axis=1)
examples = {
    "安全": capacity[security_ratio.argmin()],
    "越限": capacity[security_ratio.argmax()],
    "最近综合边界": capacity[np.abs(security_ratio - 1).argmin()],
    "最近谷值边界": capacity[np.abs(-sample[names[2]].to_numpy()-threshold[2]).argmin()],
    "原配比": [12, 20, 32], "两倍容量": [24, 40, 64],
}
rows = []
for label, s in examples.items():
    s = np.asarray(s)
    M, D = np.sum(s*m_i)/100, np.sum(s*d_i)/100
    K = s*k_i/100
    A = np.zeros((4,4))
    A[0] = np.r_[-D, 1, 1, 1]/M
    A[1:,0] = -K/tau_i
    A[1:,1:] = -np.diag(1/tau_i)
    system = ct.ss(A, [-P_check/M,0,0,0], [1,0,0,0], 0)
    end = 25 / min(-ct.poles(system).real)
    t = np.linspace(0,end,int(np.ceil(end/.002))+1)
    states = np.asarray(ct.step_response(system,T=t,return_states=True).states).reshape(4,-1)
    f = 50*states[0]
    measured = np.array([(-3*f[0]+4*f[1]-f[2])/(2*t[1]), f[-1], f.min()])
    pos = np.flatnonzero((capacity == s).all(axis=1))[0]
    err = abs(measured - sample.loc[pos,names].to_numpy(dtype=float))
    residual = states[1:,-1].sum() - P_check - D*states[0,-1]
    assert np.all(err < [1e-4, 1e-7, 2e-5]) and abs(residual)<1e-7
    rows.append([label,*s,*err,abs(residual)])
audit = pd.DataFrame(rows, columns=["工况","S1","S2","S3","RoCoF绝对误差","稳态绝对误差","谷值绝对误差","平衡残差(p.u.)"])
assert np.array_equal(sample[flags].any(axis=1), sample.any_violation)
assert not sample.loc[inertia == 1000,flags[0]].any()
assert np.all(np.diff(counts,axis=0)>=0)
assert np.all(np.isfinite(audit_values := np.array([r[4:] for r in rows])))
display(audit)
````

## 最低点频差越限的 ReLU 分类器

分类器以三台机组容量和负荷阶跃为输入。将输入缩放为

$$
\tilde{x}=\left(\frac{S_1}{100},\frac{S_2}{100},\frac{S_3}{100},P\right)^\top.
$$

单隐藏层包含 16 个 ReLU 神经元。第 $j$ 个神经元的激活值和分类器输出分数分别为

$$
h_j=\max\left(0,a_j+\sum_{i=1}^{4}w_{ij}\tilde{x}_i\right),\qquad j=1,\ldots,16,
$$

$$
z_{\rm nadir}(S_1,S_2,S_3,P)=b+\sum_{j=1}^{16}v_jh_j.
$$

其中 $w_{ij}$ 为输入层到隐藏层的权重，$a_j$ 为隐藏层偏置，$v_j$ 为第 $j$ 个神经元到输出层的权重，$b$ 为输出偏置。它们由样本训练得到。$v_j h_j$ 是该神经元对越限分数的贡献。训练结束后，这些参数固定，代入容量优化模型。优化的决策变量为 $S_1,S_2,S_3$，$P$ 为给定扰动。

输出分数经过 sigmoid 函数转换为预测越限概率：

$$
\hat p=\frac{1}{1+e^{-z_{\rm nadir}}}.
$$

$z_{\rm nadir}>0$ 时预测最低点频差越限，$z_{\rm nadir}\leq 0$ 时预测合格。动态仿真样本的真实标签以最低点频差绝对值是否严格超过 0.50 Hz 判定，等于限值为合格。

以下按 $S_1,S_2,S_3$ 各 25 MW、$P$ 每 0.25 p.u. 的联合区间分层，在每层随机分配训练集、验证集和测试集。表中括号表示各集合输入的实际最小值和最大值。

````python
# 静默当前内核反复报告的矩阵乘法和内嵌图表提示。
import warnings
warnings.filterwarnings("ignore", message=r"(divide by zero|overflow|invalid value) encountered in matmul$", category=RuntimeWarning)
warnings.filterwarnings("ignore", message=r"Consider using IPython.display.IFrame instead", category=UserWarning)

from pathlib import Path
import numpy as np
import pandas as pd
import pyarrow.parquet as pq
from sklearn.neural_network import MLPClassifier
from copy import deepcopy
from threadpoolctl import threadpool_limits
from IPython.display import display
from tqdm import tqdm
from html import escape
from io import StringIO
from IPython.display import HTML

class TrainingProgress(tqdm):
    def __init__(self, **kwargs):
        self.handle = None
        super().__init__(file=StringIO(), **kwargs)
    def display(self, msg=None, pos=None):
        view = HTML("<pre>" + escape(str(self)) + "</pre>")
        if self.handle is None:
            self.handle = display(view, display_id=True)
        else:
            self.handle.update(view)

files = sorted(Path("cw2_samples").glob("step_*.parquet"))
features = ["S1_MW","S2_MW","S3_MW","P_pu"]
assert len(files) == 100
capacity_nn = pq.read_table(files[0], columns=features[:3]).to_pandas().to_numpy(dtype=np.int16)
N = len(capacity_nn)
bins = (capacity_nn-1)//25
capacity_bin = bins @ np.array([16,4,1])
labels = np.empty((100,N), dtype=bool)
for level,file in enumerate(files):
    frame = pq.read_table(file,columns=features+["nadir_violation"]).to_pandas()
    assert np.array_equal(frame[features[:3]].values, capacity_nn)
    assert np.all(frame.P_pu == (level+1)/100)
    labels[level] = frame.nadir_violation
# 每档扰动含 25 个幅值。在各四维层内随机置换，再按累计配额切分。
split = np.empty_like(labels,dtype=np.uint8)
rng = np.random.default_rng(42)
seen = 0
for first in range(0,100,25):
    for k in np.unique(capacity_bin):
        rows = np.flatnonzero(capacity_bin==k)
        size = len(rows)*25
        # 训练和累计训练+验证的配额分别取整，最终总数精确。
        train_n = (seen+size)*7//10 - seen*7//10
        trainval_n = (seen+size)*9//10 - seen*9//10
        parts = np.full(size,2,np.uint8)
        parts[:train_n] = 0
        parts[train_n:trainval_n] = 1
        rng.shuffle(parts)
        split[first:first+25,rows] = parts.reshape(25,-1)
        seen += size
summary = []
for group,title in enumerate(["训练集","验证集","测试集"]):
    mask = split==group
    capacities = capacity_nn[mask.any(axis=0)]
    levels = np.flatnonzero(mask.any(axis=1))+1
    summary.append([title,mask.sum(),*[(c.min(),c.max()) for c in capacities.T],
                    (levels.min()/100,levels.max()/100),labels[mask].mean()])
summary = pd.DataFrame(summary,columns=["集合","样本量",*features,"最低点越限比例"])
assert list(summary["样本量"]) == [67352600,19243600,9621800]
display(summary)

# 展示送入网络的训练样本，保留原始输入和缩放后的 x。
example_level = 9  # P = 0.10 p.u.
example_rows = np.flatnonzero(split[example_level] == 0)[:5]
raw = capacity_nn[example_rows]
example = pd.DataFrame({
    "S1 (MW)": raw[:, 0], "S2 (MW)": raw[:, 1], "S3 (MW)": raw[:, 2],
    "P (p.u.)": np.full(len(raw), (example_level + 1) / 100),
    "x1 = S1/100": raw[:, 0] / 100,
    "x2 = S2/100": raw[:, 1] / 100,
    "x3 = S3/100": raw[:, 2] / 100,
    "x4 = P": np.full(len(raw), (example_level + 1) / 100),
    "y（1 越限，0 合格）": labels[example_level, example_rows].astype(int),
})
print("神经网络训练输入 x 和真实标签 y 的示例（P = 0.10 p.u.）")
display(example)
````

### ReLU 模型与训练

使用训练集分批更新网络参数，采用 Adam 优化器和二元交叉熵损失。每轮遍历全部训练样本，并将不同负荷阶跃的样本混合后输入网络。

每轮结束后计算验证集交叉熵，保存验证损失最低的一轮参数。训练最多进行 30 轮，若连续 3 轮未改善则提前停止。下表记录各轮验证集损失。

````python
# 缩放为 (S1/100,S2/100,S3/100,P)，只使用四个运行参数。
scaled_capacity = capacity_nn.astype(np.float32)/100
def inputs(level,rows):
    x = np.empty((len(rows),4),np.float32)
    x[:,:3] = scaled_capacity[rows]
    x[:,3] = (level+1)/100
    return x

def scores(model,x):
    hidden = np.maximum(x @ model.coefs_[0] + model.intercepts_[0],0)
    return (hidden @ model.coefs_[1] + model.intercepts_[1]).ravel()

# 混合不同扰动幅值后再做 Adam 更新。一个 epoch 遍历全部训练样本。
train_rows = [np.flatnonzero(s==0).astype(np.int32) for s in split]
val_rows = [np.flatnonzero(s==1).astype(np.int32) for s in split]
model = MLPClassifier(hidden_layer_sizes=(16,),activation="relu",solver="adam",
                      batch_size=2048,learning_rate_init=.002,random_state=42,
                      shuffle=False)
best_loss = np.inf
stale = 0
history = []
with threadpool_limits(limits=1), TrainingProgress(total=30,desc="训练 ReLU",unit="轮") as progress:
    for epoch in range(1,31):
        for rows in train_rows:
            rng.shuffle(rows)
        # 每次取各扰动分区的一小段，组成混合训练块。
        for start in range(0,max(map(len,train_rows)),1024):
            slices = [rows[start:start+1024] for rows in train_rows]
            x = np.concatenate([inputs(i,r) for i,r in enumerate(slices)])
            y = np.concatenate([labels[i,r] for i,r in enumerate(slices)])
            order = rng.permutation(len(x))
            model.partial_fit(x[order],y[order],classes=[False,True])
        loss = 0.
        for level,rows in enumerate(val_rows):
            z = scores(model,inputs(level,rows))
            loss += np.logaddexp(0,np.where(labels[level,rows],-z,z)).sum(dtype=float)
        loss /= sum(map(len,val_rows))
        if loss < best_loss:
            best_loss,best_epoch,best_model = loss,epoch,deepcopy(model)
            stale = 0
        else:
            stale += 1
        history.append((epoch,loss))
        progress.set_postfix(loss=loss,best_epoch=best_epoch,refresh=False)
        progress.update()
        if stale==3:
            break
display(pd.DataFrame(history,columns=["轮次","验证集交叉熵"]))
````

### 分类效果与训练参数

准确率以集合全部样本为分母。漏判表示实际越限且预测合格，漏判率以实际越限数为分母。最后一行单独检验题设工况。权重表每行对应一个隐藏神经元。

````python
def evaluate(group,levels):
    total = correct = unsafe = missed = 0
    for level in levels:
        rows = np.flatnonzero(split[level]==group)
        x = inputs(level,rows)
        true = labels[level,rows]
        pred = scores(best_model,x)>0
        # 直接以网络输出概率和显式分数核对，接近 0.5 时允许浮点舍入。
        prob = best_model.predict_proba(x)[:,1]
        assert np.all((pred==(prob>.5)) | (abs(prob-.5)<2e-6))
        total += len(rows)
        correct += np.count_nonzero(pred==true)
        unsafe += true.sum()
        missed += np.count_nonzero(true & ~pred)
    return [total,correct,correct/total,unsafe,missed,missed/unsafe]

with threadpool_limits(limits=1):
    metrics = pd.DataFrame([evaluate(1,range(100)),evaluate(2,range(100)),evaluate(2,[9])],
        index=["验证集","测试集","测试集 P=0.10"],columns=["样本数","正确分类数","准确率","实际越限数","越限漏判数","漏判率"])
display(metrics)
weights = pd.DataFrame(best_model.coefs_[0].T,columns=features)
weights["隐藏偏置 a"] = best_model.intercepts_[0]
weights["输出权重 v"] = best_model.coefs_[1].ravel()
weights.index = np.arange(1,17)
display(weights)
print("最佳轮次",best_epoch,"输出偏置 b =",float(best_model.intercepts_[1][0]))
np.savez("cw2_nadir_relu.npz",W=best_model.coefs_[0],a=best_model.intercepts_[0],
         v=best_model.coefs_[1].ravel(),b=best_model.intercepts_[1],scale=[100,100,100,1],
         features=features,best_epoch=best_epoch)


# 将完整精度的模型参数代入 z，重训后随最佳权重更新。
from IPython.display import Math
def format_number(value):
    return f"{float(value):.17g}"
def signed_term(value, symbol):
    return ("+" if value >= 0 else "-") + format_number(abs(value)) + symbol

terms = []
for j in range(16):
    inner = format_number(best_model.intercepts_[0][j])
    for value, symbol in zip(best_model.coefs_[0][:, j], [r"\frac{S_1}{100}", r"\frac{S_2}{100}", r"\frac{S_3}{100}", "P"]):
        inner += signed_term(value, symbol)
    v = best_model.coefs_[1][j, 0]
    terms.append(("+" if v >= 0 else "-") + format_number(abs(v)) + r"\max\left(0," + inner + r"\right)")
formula = r"\begin{aligned} z_{\rm nadir}={}&" + format_number(best_model.intercepts_[1][0])
for start in range(0, 16, 2):
    formula += r"\\" + "&" + "".join(terms[start:start+2])
formula += r"\end{aligned}"
display(Math(formula))
````

### 将最低点分类器嵌入容量优化

固定 $P=0.10$，代入上表的权重和偏置，要求预测类别为 0，即

$$
\boxed{z_{\rm nadir}(S_1,S_2,S_3,P)=b+\sum_{j=1}^{16}v_j\max\!\left(0,a_j+\sum_{i=1}^3w_{ij}\frac{S_i}{100}+w_{4j}P\right)\le0.}
$$

等号归入预测合格类。该式近似式（21′），与式（18′）的最小成本目标、式（19′）的惯量下限、式（20′）的准稳态频差约束及式（22′）的容量约束共同构成优化模型。预测误判按上表评估，所得容量方案仍需在动态模型中复核。

### 最低点分类约束的大 M 线性化

参考 [max 函数的线性化方法](https://blog.csdn.net/m0_43401436/article/details/107127246)，对每个神经元的 $h_j=\max\{q_j,0\}$ 分别引入两个二进制变量 $u_{j,1},u_{j,2}$。其中 $u_{j,1}=1$ 选择 $q_j$ 分支，$u_{j,2}=1$ 选择零分支。

固定题设扰动 $P=0.10$，神经元输入为

$$
q_j=a_j+w_{4j}P+\sum_{i=1}^{3}w_{ij}\frac{S_i}{100}.
$$

由容量约束 $S_i\ge0$、$\sum_iS_i\le S_{\max}$，令 $c_j=a_j+w_{4j}P$，可得到有效上下界：

$$
\begin{aligned}
L_j&=c_j+\frac{S_{\max}}{100}\min\{0,w_{1j},w_{2j},w_{3j}\},\\
U_j&=c_j+\frac{S_{\max}}{100}\max\{0,w_{1j},w_{2j},w_{3j}\}.
\end{aligned}
$$

上述上下界由容量可行域的四个顶点求得。令 $M_{j,1}=\max(0,-L_j)$、$M_{j,2}=\max(0,U_j)$，分别用于放松 $q_j$ 分支和零分支。由参考文章的最大值线性化形式，原方框约束可写为

$$
\boxed{\begin{aligned}
S_i&\ge0\quad(i=1,2,3),\qquad \sum_{i=1}^{3}S_i\le S_{\max},\\
q_j&=a_j+w_{4j}P+\sum_{i=1}^{3}w_{ij}\frac{S_i}{100},\\
h_j&\ge q_j,\qquad h_j\ge0,\\
h_j&\le q_j+M_{j,1}(1-u_{j,1}),\\
h_j&\le M_{j,2}(1-u_{j,2}),\\
u_{j,1}+u_{j,2}&\ge1,\qquad u_{j,1},u_{j,2}\in\{0,1\},\quad j=1,\ldots,16,\\
b+\sum_{j=1}^{16}v_jh_j&\le0.
\end{aligned}}
$$

当 $u_{j,1}=1$ 时，$h_j\le q_j$，结合 $h_j\ge q_j$ 得 $h_j=q_j\ge0$。当 $u_{j,2}=1$ 时，$h_j\le0$，结合 $h_j\ge0$ 得 $h_j=0$ 且 $q_j\le0$。两者都为 1 只可能发生在 $q_j=h_j=0$。因此这些约束保证 $h_j=\max\{q_j,0\}$。

这里沿用式（22′）的容量范围。若另对每台机组设置容量上限，或要求优化仅在训练样本范围内进行，应将相应界加入容量约束，并可据此重新收紧 $L_j,U_j$。


### 嵌入最低点分类器后的容量优化模型

将上节推导的大 M 约束代入式（21′），并将式（20′）乘以正的稳态响应分母，得到式（18″）至（22″）的混合整数线性模型。负荷阶跃固定为 $P=0.10$ p.u.，系统基准为 $S_{\rm B}=100$ MW。三个装机容量 $S_i$ 决定投资成本，$q_j,h_j$ 和两个分支选择变量用于表示已训练的分类器。

**最小化年化成本**

$$
\min_{S,q,h,u}\quad \sum_{i=1}^3 c_i S_i . \tag{18″}
$$

**初始 RoCoF 约束**

$$
\sum_{i=1}^3 m_i S_i\ge \frac{f_0 P S_{\rm B}}{R_{\max}}. \tag{19″}
$$

**准稳态频差约束**

$$
\sum_{i=1}^3(d_i+k_i)S_i\ge \frac{f_0 P S_{\rm B}}{F_{\rm ss,max}}. \tag{20″}
$$

**最低点频差的预测合格约束**。对每个隐藏神经元 $j=1,\ldots,16$，采用上节由式（22″）计算的常数 $M_{j,1}=\max(0,-L_j)$ 和 $M_{j,2}=\max(0,U_j)$：

$$
\left\{\begin{aligned}
q_j &= a_j+w_{4j}P+\sum_{i=1}^3 w_{ij}S_i/100,\\
h_j &\ge q_j,\qquad h_j\ge 0,\\
h_j &\le q_j+M_{j,1}(1-u_{j,1}),\\
h_j &\le M_{j,2}(1-u_{j,2}),\\
u_{j,1}+u_{j,2}&\ge 1,\qquad u_{j,1},u_{j,2}\in\{0,1\},\\
b+\sum_{j=1}^{16}v_j h_j&\le 0.
\end{aligned}\right. \tag{21″}
$$

前五行逐个神经元实施，最后一行汇总全部神经元的输出，要求分类器将该容量组合判为合格。

**装机容量约束**

$$
S_i\ge0\quad(i=1,2,3),\qquad \sum_{i=1}^3 S_i\le S_{\max}=240\ \mathrm{MW}. \tag{22″}
$$

式（19″）和式（20″）直接实施解析频率限值。式（21″）精确表示分类器的判断，所得容量方案的实际最低频率仍需代回三机动态模型复核。

式（21″）中的大 M 参数在求解前由容量约束和已训练的网络权重计算。对于每个 $j=1,\ldots,16$，令 $c_j=a_j+w_{4j}P$，神经元输入的有效上下界为

$$
\begin{aligned}
L_j&=c_j+\frac{S_{\max}}{100}\min\{0,w_{1j},w_{2j},w_{3j}\},\\
U_j&=c_j+\frac{S_{\max}}{100}\max\{0,w_{1j},w_{2j},w_{3j}\}.
\end{aligned}
$$

再计算 $M_{j,1}=\max\{0,-L_j\}$、$M_{j,2}=\max\{0,U_j\}$，将结果作为固定系数代入式（21″）。上述 $\min$ 和 $\max$ 仅用于求解前的常数计算，优化模型中的约束仍为线性形式。

### Gurobi 容量优化与扰动扫描

每次求解固定一个 $P$，重算神经元界，按式（22″）允许容量连续取值。扫描 0.01 至 1.00 p.u.，步长 0.01。表格展示最小可行扰动、题设工况及最大可行扰动。曲线仅连接有最优解的采样点。频率最低点由原模型的时域驻点与稳态值复核。

````python
import numpy as np
import pandas as pd
import gurobipy as gp
from gurobipy import GRB
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp
from IPython.display import display

with np.load("cw2_nadir_relu.npz") as saved:
    W, bias, v, b = [saved[key].astype(float) for key in ("W","a","v","b")]
costs = np.array([0.75,1.05,1.65])
inertias = np.array([14.,8.,4.])
damping = np.array([.8,1.1,1.5])
gains = np.array([15.,23.,30.])
lags = np.array([7.,3.,.8])

def solve_capacity(P):
    """给定负荷阶跃，求解式（18″）至（22″）；不可行时返回 None。"""
    offset = bias + W[3]*P
    lower = offset + 2.4*np.minimum(0,W[:3].min(axis=0))
    upper = offset + 2.4*np.maximum(0,W[:3].max(axis=0))
    with gp.Model("capacity") as model:
        model.Params.OutputFlag = 0
        model.Params.MIPGap = 1e-8
        S = model.addMVar(3,lb=0,name="S")
        q = model.addMVar(len(v),lb=lower,ub=upper,name="q")
        h = model.addMVar(len(v),lb=0,name="h")
        u = model.addMVar((2,len(v)),vtype=GRB.BINARY,name="branch")
        model.setObjective(costs @ S, GRB.MINIMIZE)
        model.addConstr(S.sum() <= 240)
        model.addConstr(inertias @ S >= 5000*P/.5)
        model.addConstr((damping+gains) @ S >= 5000*P/.3)
        model.addConstr(q == offset + W[:3].T @ S/100)
        model.addConstr(h >= q)
        model.addConstr(h <= q + np.maximum(-lower,0)*(1-u[0]))
        model.addConstr(h <= np.maximum(upper,0)*(1-u[1]))
        model.addConstr(u[0]+u[1]>=1)
        model.addConstr(v @ h + b.item() <= 0)
        model.optimize()
        if model.Status == GRB.INFEASIBLE:
            return None
        if model.Status != GRB.OPTIMAL:
            raise RuntimeError(f"P={P}, Gurobi status={model.Status}")
        return S.X.copy()

def frequency_metrics(S,P):
    M = np.sum(inertias*S)/100
    D = np.sum(damping*S)/100
    K = gains*S/100
    def dynamics(t,x):
        return np.r_[(x[1:].sum()-P-D*x[0])/M,(-K*x[0]-x[1:])/lags]
    def turning_point(t,x):
        return (x[1:].sum()-P-D*x[0])/M
    turning_point.direction = 1
    result = solve_ivp(dynamics,(0,300),np.zeros(4),events=turning_point,
                       rtol=1e-9,atol=1e-12,max_step=.25)
    if not result.success:
        raise RuntimeError(result.message)
    steady = -50*P/(D+K.sum())
    candidates = [steady,0.,*(50*x[0] for x in result.y_events[0])]
    return -50*P/M, steady, min(candidates)
````

````python
# 表中三个带符号指标均取自原三机模型。
points = np.arange(1,101)/100
solutions = [solve_capacity(p) for p in points]
capacities = np.array([s if s is not None else [np.nan]*3 for s in solutions])
annual_cost = np.sum(capacities*costs,axis=1)
valid = np.flatnonzero(np.isfinite(annual_cost))
chosen = sorted(set([valid[0],9,valid[-1]]))
rows = []
for i in chosen:
    s = solutions[i]
    if s is None:
        raise RuntimeError("代表工况不可行")
    rows.append([points[i],*s,annual_cost[i],*frequency_metrics(s,points[i])])
table = pd.DataFrame(rows,columns=["P (p.u.)","S1 (MW)","S2 (MW)","S3 (MW)","年化成本 (百万元/年)","初始 RoCoF (Hz/s)","准稳态频差 (Hz)","最低点频差 (Hz)"]).set_index("P (p.u.)")
display(table.round(6))
print("无可行解的 P（p.u.）:",points[~np.isfinite(annual_cost)].round(2).tolist())
fig,left = plt.subplots(figsize=(9,4.6))
right = left.twinx()
lines = left.plot(points,annual_cost,"k--",label="Annual cost")
for j in range(3):
    lines += right.plot(points,capacities[:,j],label=f"S{j+1}")
left.set(xlabel="Load step P (p.u.)",ylabel="Annual cost (million yuan/year)")
right.set_ylabel("Installed capacity (MW)")
left.grid(alpha=.2)
left.legend(lines,[line.get_label() for line in lines],loc="upper left")
fig.tight_layout()
plt.show()
````

### 三组代表工况的频率响应

将上表三个最优容量方案分别代入原三机动态模型。实线表示频率偏差，圆点表示有限时间内的最低点，同色虚线表示准稳态频差。

````python
# 用表中三组最优容量求解原三机动态模型。
fig, ax = plt.subplots(figsize=(9, 5))
colors = ["tab:blue", "tab:orange", "tab:green"]
time = np.linspace(0, 80, 4001)
for color, row in zip(colors, rows):
    P, s1, s2, s3 = row[:4]
    S = np.array([s1, s2, s3])
    M = np.sum(inertias * S) / 100
    D = np.sum(damping * S) / 100
    K = gains * S / 100

    def dynamics(t, x):
        return np.r_[(x[1:].sum() - P - D*x[0]) / M,
                     (-K*x[0] - x[1:]) / lags]

    def nadir_event(t, x):
        return (x[1:].sum() - P - D*x[0]) / M

    nadir_event.direction = 1
    result = solve_ivp(dynamics, (0, time[-1]), np.zeros(4), t_eval=time,
                       events=nadir_event, rtol=1e-9, atol=1e-12, max_step=0.1)
    if not result.success:
        raise RuntimeError(result.message)
    frequency = 50 * result.y[0]
    ax.plot(result.t, frequency, color=color, label=f"P = {P:.2f} p.u.")
    steady = -50*P / (D + K.sum())
    ax.axhline(steady, color=color, linestyle=":", alpha=0.7)
    if len(result.t_events[0]):
        event_frequencies = 50 * result.y_events[0][:, 0]
        lowest = np.argmin(event_frequencies)
        if event_frequencies[lowest] <= steady:
            ax.scatter(result.t_events[0][lowest], event_frequencies[lowest],
                       color=color, zorder=5)
ax.set(xlabel="Time (s)", ylabel="Frequency deviation (Hz)")
ax.grid(alpha=0.2)
ax.legend(title="Solid: response · Dotted: steady state · Dots: nadir")
fig.tight_layout()
plt.show()
````

### 频率响应随扰动的变化

三条实线由最优容量方案代入原动态模型得到，虚线表示对应的负向频率限值。无可行解的扰动幅值不绘制指标点。

````python
# 对每个可行容量方案用原动态模型计算频率指标。
responses = np.full((len(points), 3), np.nan)
for i, P in enumerate(points):
    if solutions[i] is not None:
        responses[i] = frequency_metrics(solutions[i], P)

fig, left = plt.subplots(figsize=(9, 4.8))
right = left.twinx()
lines = left.plot(points, responses[:, 0], color="tab:blue", label="Initial RoCoF")
lines += right.plot(points, responses[:, 1], color="tab:orange", label="Steady-state frequency deviation")
lines += right.plot(points, responses[:, 2], color="tab:green", label="Nadir frequency deviation")
lines += left.plot(points, np.full_like(points, -0.50), color="tab:blue", linestyle="--",
                   label="RoCoF limit (-0.50 Hz/s)")
lines += right.plot(points, np.full_like(points, -0.30), color="tab:orange", linestyle="--",
                    label="Steady-state limit (-0.30 Hz)")
lines += right.plot(points, np.full_like(points, -0.50), color="tab:green", linestyle="--",
                    label="Nadir limit (-0.50 Hz)")
left.set_xlabel("Load step P (p.u.)")
left.set_ylabel("Initial RoCoF (Hz/s)", color="tab:blue")
right.set_ylabel("Frequency deviation (Hz)")
left.grid(alpha=0.2)
left.legend(lines, [line.get_label() for line in lines], loc="center left", bbox_to_anchor=(1.12, 0.5))
fig.tight_layout()
plt.show()
````
